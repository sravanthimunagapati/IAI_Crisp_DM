## Business Understanding

**CRISP-DM Phase 1.** This is the planning and framing stage that precedes any data work, presented in full on Slide 2 ("Automating the Inspection Decision"). Summarised here for traceability.

**Current process.** Inject & hold → part ejected → IR camera scan → scale (weight) → vision camera → human inspector assigns the defect label. The human inspector step is the bottleneck: subjective, operator-dependent, and does not scale with machine throughput.

**Business problem.** Injection moulding produces plastic parts at high speed and volume. Defects occur when process parameters drift. A human inspector currently examines every part after ejection — inconsistent across operators and unable to keep pace with production volume.

**What we solved.** Predict which specific defect type is present per cycle — not just OK or NOK — using sensor data already collected automatically on the production line, with no new instrumentation required.

**Success criteria — the triage layer.** The model does not need to replace the inspector outright to be useful. Sensor data feeds the model; the model's output routes each part one of three ways: all classifiers negative → part passes automatically; high-confidence defect → auto-flagged; borderline → routed to human review. Success is human review reduced to the genuinely uncertain cases only, not full automation.

## Data Understanding

**CRISP-DM Phase 2.** Before any cleaning or modelling, we examine the dataset as delivered: its shape, its grouping structure, and the raw sensor signals themselves.

— Load the ProBayes parquet file; confirm row/column counts, material split, and the raw experiment identifier.
— Audit the experiment grouping structure. This is where we catch that `MET_ExperimentNumber` repeats across materials (e.g. `A01` exists for both PP and ABS as two distinct conditions) and construct the corrected `EXP_GROUP` compound key used for all cross-validation from here on.

— Audit all 334 raw columns by source group (E77, SET, ENV, IR, CV, SCA, DXP, SIM, etc.) and missingness pattern, to inform what gets kept in Data Preparation.

— Inspect the raw DXP pressure waveforms directly: signal availability, array lengths, and the three visible phases (injection spike, holding plateau, cooling decay) for one sample cycle.

No targets are touched yet — this phase is purely about understanding what we have and where the structural risks are (the EXP_GROUP leakage risk found here shapes the entire evaluation strategy in Modelling and Evaluation).

```python
# ============================================================
# Cell 1: Imports & Data Loading
# ============================================================

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# Load dataset
df = pd.read_parquet("fk2gn-x8e76/dataset.parquet")

# ── Basic shape ──────────────────────────────────────────────
print(f"Dataset shape: {df.shape}")
print(f"Rows (cycles): {len(df)}")
print(f"Columns (features): {len(df.columns)}")

# ── Material split ───────────────────────────────────────────
print("\n── Material split ──")
print(df['MET_MaterialName'].value_counts())

# ── Experiment groups ────────────────────────────────────────
print(f"\n── Experiment groups ──")
print(f"Unique experiments: {df['MET_ExperimentNumber'].nunique()}")
print(f"Parts per experiment (should be 12): ")
print(df['MET_ExperimentNumber'].value_counts().describe())

# ── Column prefixes present ──────────────────────────────────
prefixes = ['MET','ENV','E77','SET','DXP','TCE','TCN','IR','CV','SCA','LBL','SIM','DOS','DRY']
print("\n── Columns per source ──")
for p in prefixes:
    cols = [c for c in df.columns if c.startswith(p+'_')]
    if cols:
        print(f"  {p:<5} {len(cols):>3} columns")

# ── Target labels ────────────────────────────────────────────
TARGET_BINARY = ['LBL_SinkMarks', 'LBL_SprueCircle', 'LBL_Underfilled']
TARGET_STREAKS = ['LBL_StreaksLevel1', 'LBL_StreaksLevel2', 'LBL_StreaksLevel3']

print("\n── Label distributions ──")
for col in TARGET_BINARY + TARGET_STREAKS:
    pos = int(df[col].sum())
    pct = 100 * pos / len(df)
    print(f"  {col:<25} {pos:>3}/{len(df)}  ({pct:.1f}% positive)")

# ── Streaks ordinal target preview ───────────────────────────
# Collapse three streak columns into one ordinal: 0/1/2/3
df['LBL_Streaks'] = (
    df['LBL_StreaksLevel1'] * 1 +
    df['LBL_StreaksLevel2'] * 2 +
    df['LBL_StreaksLevel3'] * 3
).clip(0, 3).astype(int)

print("\n── Streaks ordinal distribution ──")
print(df['LBL_Streaks'].value_counts().sort_index())

# ── DXP column types — are they arrays or scalars? ───────────
dxp_cols = [c for c in df.columns if c.startswith('DXP_')]
print(f"\n── DXP columns: {len(dxp_cols)} total ──")
sample_dxp = df[dxp_cols[0]].iloc[0]
print(f"  DXP sample type: {type(sample_dxp)}")
print(f"  DXP sample length (if array): {len(sample_dxp) if hasattr(sample_dxp, '__len__') else 'scalar'}")
```

```python
# ============================================================
# Cell 2: Experiment Group Structure Audit
# DIAGNOSTIC: MET_ExperimentNumber repeats across materials/dates.
# Purpose: find correct grouping key for GroupKFold.
# Finding: must use ExperimentNumber+Material+Date = 47 unique groups.
# ============================================================

exp_material = df.groupby(['MET_ExperimentNumber', 'MET_MaterialName']).size().unstack(fill_value=0)
print(exp_material)

# ── How do experiment numbers map across dates ──
df['MET_Date'] = pd.to_datetime(df['MET_Timestamp']).dt.date
print("\n── Experiment numbers × Date ──")
exp_date = df.groupby(['MET_ExperimentNumber', 'MET_Date']).size().unstack(fill_value=0)
print(exp_date)

# ── Are experiment numbers unique across material+date combos? ──
print("\n── Unique experiment ID = ExperimentNumber + Material + Date ──")
df['EXP_GROUP'] = (
    df['MET_ExperimentNumber'].astype(str) + '_' +
    df['MET_MaterialName'].astype(str) + '_' +
    df['MET_Date'].astype(str)
)
print(f"Unique MET_ExperimentNumber alone:     {df['MET_ExperimentNumber'].nunique()}")
print(f"Unique ExperimentNumber + Material:    {(df['MET_ExperimentNumber'].astype(str) + '_' + df['MET_MaterialName']).nunique()}")
print(f"Unique ExperimentNumber + Mat + Date:  {df['EXP_GROUP'].nunique()}")

# ── Parts per unique group ──
print("\n── Parts per EXP_GROUP ──")
print(df['EXP_GROUP'].value_counts().describe())
print("\nGroups with != 12 parts:")
odd = df['EXP_GROUP'].value_counts()
print(odd[odd != 12])

# Correct grouping key for GroupKFold
df['EXP_GROUP'] = (
    df['MET_ExperimentNumber'].astype(str) + '_' +
    df['MET_MaterialName'].astype(str) + '_' +
    df['MET_Date'].astype(str)
)
```

```python
# ============================================================
# Cell 3: Feature Audit
# DIAGNOSTIC: classify all 334 columns — scalar, timeseries, string.
# Finding: 72.3% missing = March-only params (structural, not random);
# 27.7% missing = IR Img3 (March had 2 images only); SIM excluded per prof.
# ============================================================

# ── Classify every column by type ────────────────────────────
scalar_cols = []
timeseries_cols = []
string_cols = []
drop_cols = []

for col in df.columns:
    val = df[col].dropna().iloc[0] if df[col].notna().any() else None
    if val is None:
        drop_cols.append((col, 'all null'))
    elif isinstance(val, (np.ndarray, list)):
        timeseries_cols.append(col)
    elif isinstance(val, str):
        string_cols.append(col)
    else:
        scalar_cols.append(col)

print(f"Scalar columns:      {len(scalar_cols)}")
print(f"Timeseries columns:  {len(timeseries_cols)}")
print(f"String columns:      {len(string_cols)}")
print(f"All-null columns:    {len(drop_cols)}")

# ── String columns — what are they? ──────────────────────────
print("\n── String columns ──")
for col in string_cols:
    uniq = df[col].nunique()
    print(f"  {col:<35} {uniq} unique values")

# ── Timeseries columns — which sources? ──────────────────────
print("\n── Timeseries columns by source ──")
for prefix in ['DXP', 'TCE', 'TCN', 'DOS', 'DRY']:
    ts = [c for c in timeseries_cols if c.startswith(prefix+'_')]
    if ts:
        # Sample length from first non-null value
        sample = df[ts[0]].dropna().iloc[0]
        print(f"  {prefix:<5} {len(ts):>3} columns  |  length per cycle: {len(sample)}")

# ── Scalar columns by source ──────────────────────────────────
print("\n── Scalar columns by source ──")
for prefix in ['MET','ENV','E77','SET','IR','CV','SCA','LBL','SIM']:
    sc = [c for c in scalar_cols if c.startswith(prefix+'_')]
    if sc:
        print(f"  {prefix:<5} {len(sc):>3} columns")

# ── Missing values in scalar columns ─────────────────────────
print("\n── Missing values in scalar features (>0% only) ──")
scalar_df = df[scalar_cols]
missing = scalar_df.isnull().mean() * 100
missing = missing[missing > 0].sort_values(ascending=False)
if len(missing) == 0:
    print("  No missing values in scalar columns")
else:
    for col, pct in missing.items():
        prefix = col.split('_')[0]
        print(f"  {col:<40} {pct:.1f}% missing  [{prefix}]")

# ── SIM columns — confirm present and list them ───────────────
sim_cols = [c for c in df.columns if c.startswith('SIM_')]
print(f"\n── SIM columns to exclude: {len(sim_cols)} ──")
print([c for c in sim_cols])

# ── Feature groups we will USE ────────────────────────────────
print("\n── Feature groups for modelling ──")
use_scalar = ['E77', 'SET', 'ENV', 'IR', 'CV', 'SCA']
use_ts     = ['DXP']
exclude    = ['SIM', 'LBL', 'MET', 'DOS', 'DRY', 'TCE', 'TCN']

print("  INCLUDE (scalar):", use_scalar)
print("  INCLUDE (timeseries):", use_ts)
print("  EXCLUDE:", exclude)
print("\n  Note: DOS/DRY/TCE/TCN excluded for now — timeseries,")
print("        low expected signal vs complexity added.")
print("        TCE/TCN may be revisited if DXP alone is weak.")
```

## Data Preparation

**CRISP-DM Phase 3.** Based on the Data Understanding audit, we now build the clean feature matrix that modelling will actually use.

 — Select feature columns by source group, dropping anything with structural missingness (e.g. SET parameters only present for March 2022, IR Image 3 only present for December 2021) and anything not numeric or not available at inference time (SIM columns excluded — simulation data doesn't exist in production).
 
— Apply variance and correlation filters to the remaining scalar features (334 → 63), removing near-constant and redundant columns.

— Extract 35 phase-based scalar features from the raw DXP pressure curves, using trigger arrays as phase masks (injection / holding / gate freeze / cooling).

— Sanity-check the DXP-derived features: confirm they actually separate OK from defective parts before trusting them as model inputs (e.g. holding cavity pressure is ~75% lower in SinkMarks-defective parts).

By the end of this phase we have two parallel feature representations ready for modelling: a 63-column scalar matrix, and the raw two-channel DXP waveform array (built later in Cells 11–12 specifically for the TCN).

```python
# ============================================================
# Cell 4: Feature Selection and Clean Feature Matrix
# ============================================================

SET_COLS = [
    'SET_CylinderTemperature', 'SET_ToolTemperature',
    'SET_HoldingPressure1', 'SET_HoldingPressure2'
]

# EUROMAP77 scalars — drop those with 27.7% structural missing
# (only present for Dec 2021, not March 2022)
e77_missing = [
    'E77_HoldingPressureMax', 'E77_InjectionSpeed1', 'E77_InjectionSpeed2',
    'E77_MeltPressure2Max', 'E77_PlastificationSpeed'
]
E77_COLS = [c for c in df.columns
            if c.startswith('E77_')
            and c not in e77_missing
            and df[c].dtype in [np.float64, np.int64, float, int]]

# ENV — all present
ENV_COLS = ['ENV_AirTemperature', 'ENV_AirHumidity']

# IR — Img1 and Img2 only (Img3 missing 27.7%)
IR_COLS = [c for c in df.columns
           if c.startswith('IR_')
           and 'Img3' not in c
           and df[c].dtype in [np.float64, np.int64, float, int]]

# CV — drop CV_Width (72.3% missing), drop image name strings
CV_COLS = [c for c in df.columns
           if c.startswith('CV_')
           and c != 'CV_Width'
           and df[c].dtype in [np.float64, np.int64, float, int]]

# SCA — part weight
SCA_COLS = ['SCA_PartWeight']

# Material — encode as binary
MAT_COL = ['MET_MaterialName']

# DXP raw timeseries — kept separately for feature extraction in Cell 5
DXP_TS_COLS = [c for c in df.columns
               if c.startswith('DXP_')
               and df[c].dropna().iloc[0].__class__.__name__ == 'ndarray']

# ── Report what we have ───────────────────────────────────────
print("── Feature groups selected ──")
print(f"  SET    {len(SET_COLS):>3} columns  {SET_COLS}")
print(f"  E77    {len(E77_COLS):>3} columns")
print(f"  ENV    {len(ENV_COLS):>3} columns")
print(f"  IR     {len(IR_COLS):>3} columns  (Img1 + Img2 only)")
print(f"  CV     {len(CV_COLS):>3} columns  (CV_Width dropped)")
print(f"  SCA    {len(SCA_COLS):>3} columns")
print(f"  MAT      1 column   (encoded)")
print(f"  DXP_TS {len(DXP_TS_COLS):>3} timeseries columns (for Cell 5)")

SCALAR_FEATURES = SET_COLS + E77_COLS + ENV_COLS + IR_COLS + CV_COLS + SCA_COLS

print(f"\n  Total scalar features: {len(SCALAR_FEATURES)}")
print(f"  Total DXP timeseries:  {len(DXP_TS_COLS)}")

# ── Build scalar feature matrix ───────────────────────────────
from sklearn.preprocessing import LabelEncoder

X = df[SCALAR_FEATURES + MAT_COL].copy()

# Encode material as binary (PP=1, ABS=0 or similar)
le = LabelEncoder()
X['MET_MaterialName'] = le.fit_transform(X['MET_MaterialName'])
print(f"\n  Material encoding: {dict(zip(le.classes_, le.transform(le.classes_)))}")

# ── Impute missing values ─────────────────────────────────────
# Strategy: median per material group
# Reason: E77 missingness is structural by date/material,
#         so within-material median is the honest fill
material_col = X['MET_MaterialName'].copy()

for col in X.columns:
    if X[col].isnull().any():
        X[col] = X.groupby(material_col)[col].transform(
            lambda s: s.fillna(s.median())
        )

print(f"\n── Missing values after imputation ──")
remaining = X.isnull().sum().sum()
print(f"  Total remaining NaN: {remaining}")

# ── Build target matrix ───────────────────────────────────────
TARGET_BINARY = ['LBL_SinkMarks', 'LBL_SprueCircle', 'LBL_Underfilled']

# Streaks as single ordinal target 0/1/2/3
# Already created in Cell 1 as df['LBL_Streaks']
y_binary = df[TARGET_BINARY].fillna(0).astype(int)
y_streaks = df['LBL_Streaks'].astype(int)

print(f"\n── Target shapes ──")
print(f"  Binary targets:  {y_binary.shape}  {TARGET_BINARY}")
print(f"  Streaks target:  {y_streaks.shape}  (0/1/2/3 ordinal)")

print(f"\n── Final binary label distributions ──")
for col in TARGET_BINARY:
    pos = int(y_binary[col].sum())
    pct = 100 * pos / len(y_binary)
    print(f"  {col:<25} {pos:>3}/{len(y_binary)}  ({pct:.1f}% positive)")

print(f"\n── Streaks ordinal distribution ──")
print(y_streaks.value_counts().sort_index())

# ── GroupKFold setup ──────────────────────────────────────────
groups = df['EXP_GROUP'].values
print(f"\n── GroupKFold setup ──")
print(f"  Grouping key: EXP_GROUP")
print(f"  Unique groups: {len(np.unique(groups))}")
print(f"  Feature matrix shape: {X.shape}")
```

```python
# ============================================================
# Cell 5: Remove Low-Variance and Highly Correlated Features
# ENGINEERING FILTER: remove noise before modelling.
# Variance < 0.01 dropped (11 features); correlation > 0.95 dropped (31).
# Result: 105 -> 63 features. Better ratio for 564 samples.
# ============================================================

from sklearn.feature_selection import VarianceThreshold

print("── Before filtering ──")
print(f"  Features: {X.shape[1]}")

# ── Step 1: Remove near-zero variance features ───────────────
# Threshold: features that are essentially constant across cycles
# Using 0.01 — a feature must have at least 1% variance to stay
selector = VarianceThreshold(threshold=0.01)
selector.fit(X)
low_var_mask = selector.get_support()
dropped_var = [col for col, keep in zip(X.columns, low_var_mask) if not keep]

print(f"\n── Low variance features dropped ({len(dropped_var)}) ──")
for col in dropped_var:
    print(f"  {col:<45} var={X[col].var():.6f}")

X = X[X.columns[low_var_mask]]

# ── Step 2: Remove highly correlated features ────────────────
# Keep one from each pair with correlation > 0.95
# Priority: keep the one with higher variance (more informative)
corr_matrix = X.corr().abs()
upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

dropped_corr = []
for col in upper.columns:
    if any(upper[col] > 0.95):
        # Find which column it's correlated with
        correlated_with = upper.index[upper[col] > 0.95].tolist()
        # Drop the one with lower variance
        candidates = [col] + correlated_with
        variances = {c: X[c].var() for c in candidates if c in X.columns}
        if variances:
            to_drop = min(variances, key=variances.get)
            if to_drop in X.columns and to_drop not in dropped_corr:
                dropped_corr.append(to_drop)

print(f"\n── Highly correlated features dropped (>{0.95}) : {len(dropped_corr)} ──")
for col in dropped_corr[:20]:  # show first 20
    print(f"  {col}")
if len(dropped_corr) > 20:
    print(f"  ... and {len(dropped_corr)-20} more")

X = X.drop(columns=[c for c in dropped_corr if c in X.columns])

print(f"\n── After filtering ──")
print(f"  Features remaining: {X.shape[1]}")
print(f"  Dropped (low variance): {len(dropped_var)}")
print(f"  Dropped (high correlation): {len(dropped_corr)}")
print(f"  Total dropped: {len(dropped_var) + len(dropped_corr)}")

# ── Show what survived by source ─────────────────────────────
print(f"\n── Surviving features by source ──")
for prefix in ['SET','E77','ENV','IR','CV','SCA','MET']:
    cols = [c for c in X.columns if c.startswith(prefix+'_') or c == 'MET_MaterialName']
    if prefix == 'MET':
        cols = [c for c in X.columns if c == 'MET_MaterialName']
    if cols:
        print(f"  {prefix:<5} {len(cols):>3} remaining")
```

```python
# ============================================================
# Cell 6: DXP Timeseries Signal Inspection
# DIAGNOSTIC: inspect raw DXP waveforms before feature extraction.
# Finding: all signals 100% present; lengths 8041-12079 timesteps;
# 3 clear phases visible: injection spike, holding plateau, cooling decay.
# ============================================================

dxp_triggers = [c for c in DXP_TS_COLS if 'Trig' in c]
dxp_signals  = [c for c in DXP_TS_COLS if 'Trig' not in c]

print(f"DXP trigger columns:   {len(dxp_triggers)}")
print(f"DXP signal columns:    {len(dxp_signals)}")

print(f"\n── Trigger columns ──")
for c in dxp_triggers:
    print(f"  {c}")

print(f"\n── Signal columns ──")
for c in dxp_signals:
    print(f"  {c}")

# ── Check signal availability and length consistency ─────────
print("\n── Signal availability and array lengths ──")
for col in dxp_signals:
    pct = df[col].notna().mean() * 100
    lengths = df[col].dropna().apply(len)
    print(f"  {col:<35} {pct:.0f}% present  "
          f"min_len={lengths.min()}  max_len={lengths.max()}")

# ── Plot one cycle of the key pressure signals ───────────────
import matplotlib.pyplot as plt

sample_idx = df[df['EXP_GROUP'].str.startswith('Z01_PP')].index[0]

key_signals = [c for c in dxp_signals if any(
    kw in c for kw in ['PrsAct', 'Inj1Prs', 'CavPrs', 'ScrwVel', 'ScrwPos']
)]

if not key_signals:
    key_signals = dxp_signals[:4]

print(f"\n── Plotting signals for cycle: {sample_idx} "
      f"(EXP_GROUP={df.loc[sample_idx,'EXP_GROUP']}) ──")

fig, axes = plt.subplots(len(key_signals), 1,
                          figsize=(14, 3*len(key_signals)),
                          sharex=True)
if len(key_signals) == 1:
    axes = [axes]

for ax, col in zip(axes, key_signals):
    signal = df.loc[sample_idx, col]
    time   = np.arange(len(signal)) * 0.005
    ax.plot(time, signal, linewidth=0.8)
    ax.set_ylabel(col.replace('DXP_',''), fontsize=8)
    ax.grid(True, alpha=0.3)

axes[-1].set_xlabel('Time (seconds)')
fig.suptitle(f'DXP Signals — One Cycle (Z01_PP, optimal settings)',
             fontsize=11)
plt.tight_layout()
plt.savefig('dxp_signals_sample.png', dpi=120, bbox_inches='tight')
plt.show()
print("Saved: dxp_signals_sample.png")
```

```python
# ============================================================
# Cell 7: DXP Feature Extraction from Phase-Masked Signals
# ============================================================

def extract_dxp_features(row):
    """
    Extract scalar features from raw DXP timeseries for one cycle.
    Uses trigger columns as phase masks.
    """
    features = {}

    # ── Load signals ─────────────────────────────────────────
    inj_prs  = np.array(row['DXP_Inj1PrsAct'],    dtype=np.float32)
    cav_prs  = np.array(row['DXP_MldCavPrs1Act'],  dtype=np.float32)

    # ── Load trigger masks ───────────────────────────────────
    trig_inj  = np.array(row['DXP_TrigInj1'],  dtype=bool)
    trig_hld  = np.array(row['DXP_TrigHld1'],  dtype=bool)
    trig_cool = np.array(row['DXP_TrigCool'],   dtype=bool)
    trig_plst = np.array(row['DXP_TrigPlst1'],  dtype=bool)

    dt = 0.005  # seconds per timestep

    # ── Helper: safe stats on a masked segment ───────────────
    def seg_stats(signal, mask, prefix):
        s = signal[mask] if mask.sum() > 0 else np.array([0.0])
        features[f'{prefix}_max']    = float(np.max(s))
        features[f'{prefix}_mean']   = float(np.mean(s))
        features[f'{prefix}_std']    = float(np.std(s))
        features[f'{prefix}_area']   = float(np.sum(s) * dt)
        features[f'{prefix}_duration'] = float(mask.sum() * dt)

    # ── INJECTION PHASE — Inj1PrsAct ────────────────────────
    seg_stats(inj_prs, trig_inj, 'INJ_InjPrs')

    # Rise rate: slope from start of injection to peak
    if trig_inj.sum() > 1:
        inj_seg = inj_prs[trig_inj]
        peak_idx = np.argmax(inj_seg)
        if peak_idx > 0:
            features['INJ_InjPrs_rise_rate'] = float(
                inj_seg[peak_idx] / (peak_idx * dt)
            )
        else:
            features['INJ_InjPrs_rise_rate'] = 0.0
        features['INJ_InjPrs_peak_time'] = float(peak_idx * dt)
    else:
        features['INJ_InjPrs_rise_rate'] = 0.0
        features['INJ_InjPrs_peak_time'] = 0.0

    # ── HOLDING PHASE — Inj1PrsAct ──────────────────────────
    seg_stats(inj_prs, trig_hld, 'HLD_InjPrs')

    # Holding stability: coefficient of variation
    hld_mean = features['HLD_InjPrs_mean']
    hld_std  = features['HLD_InjPrs_std']
    features['HLD_InjPrs_cv'] = float(
        hld_std / hld_mean if hld_mean > 0 else 0.0
    )

    # ── INJECTION PHASE — Cavity Pressure ───────────────────
    seg_stats(cav_prs, trig_inj, 'INJ_CavPrs')

    # ── HOLDING PHASE — Cavity Pressure ─────────────────────
    seg_stats(cav_prs, trig_hld, 'HLD_CavPrs')

    # ── GATE FREEZE — when cavity pressure drops sharply ────
    # Look in the transition between holding and cooling
    # Gate freeze = point of steepest negative slope in cav_prs
    # during end of holding phase
    if trig_hld.sum() > 10:
        cav_hld_seg = cav_prs[trig_hld]
        # Gradient of cavity pressure during holding
        grad = np.gradient(cav_hld_seg)
        freeze_idx = np.argmin(grad)  # steepest drop
        features['GATE_freeze_pressure'] = float(cav_hld_seg[freeze_idx])
        features['GATE_freeze_time']     = float(freeze_idx * dt)
        features['GATE_freeze_drop_rate'] = float(grad[freeze_idx])
    else:
        features['GATE_freeze_pressure']  = 0.0
        features['GATE_freeze_time']      = 0.0
        features['GATE_freeze_drop_rate'] = 0.0

    # ── COOLING PHASE — Cavity Pressure decay ────────────────
    if trig_cool.sum() > 1:
        cav_cool = cav_prs[trig_cool]
        features['COOL_CavPrs_start'] = float(cav_cool[0])
        features['COOL_CavPrs_end']   = float(cav_cool[-1])
        features['COOL_CavPrs_decay'] = float(
            (cav_cool[0] - cav_cool[-1]) / (trig_cool.sum() * dt)
            if trig_cool.sum() > 0 else 0.0
        )
        features['COOL_duration'] = float(trig_cool.sum() * dt)
    else:
        features['COOL_CavPrs_start'] = 0.0
        features['COOL_CavPrs_end']   = 0.0
        features['COOL_CavPrs_decay'] = 0.0
        features['COOL_duration']     = 0.0

    # ── PLASTIFICATION — screw torque proxy ─────────────────
    seg_stats(inj_prs, trig_plst, 'PLST_InjPrs')

    return features

# ── Apply to all 564 cycles ──────────────────────────────────
print("Extracting DXP features for all cycles...")
dxp_feature_rows = df.apply(extract_dxp_features, axis=1)
X_dxp = pd.DataFrame(list(dxp_feature_rows))
X_dxp.index = df.index

print(f"DXP features extracted: {X_dxp.shape}")
print(f"\n── Feature names ──")
for col in X_dxp.columns:
    print(f"  {col}")

# ── Check for any NaN in extracted features ──────────────────
nan_count = X_dxp.isnull().sum().sum()
print(f"\nNaN in DXP features: {nan_count}")

# ── Quick sanity check — do features vary across experiments?─
print(f"\n── Feature variance check (std across all cycles) ──")
low_var_dxp = X_dxp.std()[X_dxp.std() < 0.01]
if len(low_var_dxp) > 0:
    print(f"  Low variance DXP features: {list(low_var_dxp.index)}")
else:
    print(f"  All DXP features show meaningful variance. Good.")

# ── Merge DXP features into main feature matrix ──────────────
X_combined = pd.concat([X, X_dxp], axis=1)
print(f"\n── Combined feature matrix ──")
print(f"  Scalar only:    {X.shape[1]} features")
print(f"  DXP extracted:  {X_dxp.shape[1]} features")
print(f"  Combined total: {X_combined.shape[1]} features")
```

## Modelling — Part 1: Scalar Baseline and Feature-Set Comparison

**CRISP-DM Phase 4.** With clean features in hand, we establish a baseline model and use it to settle an open question from Data Preparation: do the raw DXP curves add anything beyond what's already captured in the 63 scalar features?

— Train the LightGBM baseline under GroupKFold(5), grouped by `EXP_GROUP`. This is also where the random-split vs. GroupKFold comparison is first run, motivated directly by the EXP_GROUP leakage risk identified in Data Understanding.

— Diagnose fold-to-fold variance and run the three-way feature-set comparison (scalar-only / combined / DXP-only) that determines, per target, whether the raw pressure curve carries independent signal. This is the evidence behind treating Underfilled differently from the other five defects.

```python
# ============================================================
# Cell 9: LightGBM Baseline — GroupKFold Evaluation
# ============================================================

from lightgbm import LGBMClassifier
from sklearn.model_selection import GroupKFold
from sklearn.metrics import f1_score, classification_report
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

gkf    = GroupKFold(n_splits=5)
groups = df['EXP_GROUP'].values

BINARY_TARGETS = ['LBL_SinkMarks', 'LBL_SprueCircle', 'LBL_Underfilled']

def run_groupkfold(X_input, label, y_series, is_multiclass=False):
    """
    Run GroupKFold CV for one target.
    Returns mean F1-macro across folds.
    """
    fold_f1s = []
    fold_reports = []

    for fold, (train_idx, test_idx) in enumerate(
            gkf.split(X_input, y_series, groups)):

        X_tr, X_te = X_input.iloc[train_idx], X_input.iloc[test_idx]
        y_tr, y_te = y_series.iloc[train_idx], y_series.iloc[test_idx]

        if is_multiclass:
            clf = LGBMClassifier(
                n_estimators=300,
                learning_rate=0.05,
                num_leaves=31,
                class_weight='balanced',
                objective='multiclass',
                num_class=4,
                random_state=42,
                verbose=-1
            )
            average = 'macro'
        else:
            clf = LGBMClassifier(
                n_estimators=300,
                learning_rate=0.05,
                num_leaves=31,
                class_weight='balanced',
                objective='binary',
                random_state=42,
                verbose=-1
            )
            average = 'macro'

        clf.fit(X_tr, y_tr)
        y_pred = clf.predict(X_te)
        f1 = f1_score(y_te, y_pred, average=average, zero_division=0)
        fold_f1s.append(f1)
        fold_reports.append((y_te, y_pred))

    mean_f1 = np.mean(fold_f1s)
    std_f1  = np.std(fold_f1s)
    return mean_f1, std_f1, fold_f1s, fold_reports

# ── Run A: Scalar features only ──────────────────────────────
print("=" * 60)
print("RUN A — Scalar features only")
print("=" * 60)
results_A = {}
for target in BINARY_TARGETS:
    mean_f1, std_f1, folds, _ = run_groupkfold(
        X, y_binary[target], y_binary[target]
    )
    results_A[target] = (mean_f1, std_f1, folds)
    print(f"  {target:<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
          f"folds={[f'{f:.2f}' for f in folds]}")

mean_f1, std_f1, folds, _ = run_groupkfold(
    X, y_streaks, y_streaks, is_multiclass=True
)
results_A['LBL_Streaks'] = (mean_f1, std_f1, folds)
print(f"  {'LBL_Streaks':<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
      f"folds={[f'{f:.2f}' for f in folds]}")

# ── Run B: Scalar + DXP features ─────────────────────────────
print("\n" + "=" * 60)
print("RUN B — Scalar + DXP extracted features")
print("=" * 60)
results_B = {}
for target in BINARY_TARGETS:
    mean_f1, std_f1, folds, _ = run_groupkfold(
        X_combined, y_binary[target], y_binary[target]
    )
    results_B[target] = (mean_f1, std_f1, folds)
    print(f"  {target:<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
          f"folds={[f'{f:.2f}' for f in folds]}")

mean_f1, std_f1, folds, _ = run_groupkfold(
    X_combined, y_streaks, y_streaks, is_multiclass=True
)
results_B['LBL_Streaks'] = (mean_f1, std_f1, folds)
print(f"  {'LBL_Streaks':<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
      f"folds={[f'{f:.2f}' for f in folds]}")

# ── Delta: did DXP features help? ────────────────────────────
print("\n" + "=" * 60)
print("DELTA — Does adding DXP features improve F1?")
print("=" * 60)
print(f"  {'Target':<25} {'Run A':>8} {'Run B':>8} {'Delta':>8}")
print("-" * 55)
for target in BINARY_TARGETS + ['LBL_Streaks']:
    a = results_A[target][0]
    b = results_B[target][0]
    delta = b - a
    direction = '↑' if delta > 0.01 else ('↓' if delta < -0.01 else '→')
    print(f"  {target:<25} {a:>8.3f} {b:>8.3f} {delta:>+8.3f} {direction}")
```

```python
# ============================================================
# Cell 10: Fold Variance Diagnosis and DXP-Only Run
# DIAGNOSTIC: investigate why DXP combined features hurt performance.
# Checks: fold composition (material/date balance) + DXP features alone.
# Finding: fold variance is genuine, not imbalance; DXP-only beats scalar
# only for Underfilled — E77 already captures equivalent signal elsewhere.
# Decision: target-specific feature sets, evidence-driven not assumed.
# ============================================================

for fold, (train_idx, test_idx) in enumerate(
        gkf.split(X, y_binary['LBL_SinkMarks'], groups)):
    test_groups = np.unique(groups[test_idx])
    # Which materials in this fold?
    materials = df.iloc[test_idx]['MET_MaterialName'].unique()
    dates     = df.iloc[test_idx]['MET_Date'].unique()
    print(f"Fold {fold+1}:")
    print(f"  Groups:    {list(test_groups)}")
    print(f"  Materials: {list(materials)}")
    print(f"  Dates:     {list(dates)}")
    print(f"  N parts:   {len(test_idx)}")
    print()

# ── DXP features only run ────────────────────────────────────
print("=" * 60)
print("RUN C — DXP features only")
print("=" * 60)
for target in BINARY_TARGETS:
    mean_f1, std_f1, folds, _ = run_groupkfold(
        X_dxp, y_binary[target], y_binary[target]
    )
    print(f"  {target:<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
          f"folds={[f'{f:.2f}' for f in folds]}")

mean_f1, std_f1, folds, _ = run_groupkfold(
    X_dxp, y_streaks, y_streaks, is_multiclass=True
)
print(f"  {'LBL_Streaks':<25} F1={mean_f1:.3f} ± {std_f1:.3f}  "
      f"folds={[f'{f:.2f}' for f in folds]}")
```

## Modelling — Part 2: TCN for Underfilled

**CRISP-DM Phase 4 (continued).** The comparison showed DXP-only features uniquely outperform scalar features for Underfilled. This block builds and trains the deep learning model that exploits that finding.

 — Verify the phase-cropping strategy for the raw DXP waveform before committing to an input shape for the network.

— Build the cropped/padded TCN input array and define the dilated causal convolution architecture.

— Train and evaluate the TCN under GroupKFold(5), using the same grouping key as the LightGBM models for a fair, consistent evaluation across the whole pipeline.

— Nested GroupKFold hyperparameter tuning for the LightGBM models (outer loop for performance estimation, inner loop for parameter selection, never touching the test fold).

```python
# ============================================================
# Cell 11: TCN Input Preparation — Phase Crop Verification
# DIAGNOSTIC: verify phase extraction lengths before building TCN input.
# Finding: min=3539, max=6754 timesteps; no missing phases;
# crop to median 5331 (26.7s) balances truncation vs padding.
# ============================================================

phase_lengths = []
issues = []

for idx, row in df.iterrows():
    trig_inj = np.array(row['DXP_TrigInj1'], dtype=bool)
    trig_hld = np.array(row['DXP_TrigHld1'], dtype=bool)

    # Combined phase mask
    phase_mask = trig_inj | trig_hld

    phase_len = phase_mask.sum()
    phase_lengths.append(phase_len)

    # Flag if phase is suspiciously short or missing
    if phase_len < 100:
        issues.append((idx, phase_len, row['EXP_GROUP']))

phase_lengths = np.array(phase_lengths)

print("── Injection + Holding phase lengths (timesteps) ──")
print(f"  Min:    {phase_lengths.min()}")
print(f"  Max:    {phase_lengths.max()}")
print(f"  Mean:   {phase_lengths.mean():.0f}")
print(f"  Std:    {phase_lengths.std():.0f}")
print(f"  Median: {np.median(phase_lengths):.0f}")

print(f"\n── Distribution of phase lengths ──")
bins = [0, 500, 1000, 2000, 3000, 4000, 5000, 99999]
labels = ['<500','500-1k','1k-2k','2k-3k','3k-4k','4k-5k','>5k']
for i, (lo, hi) in enumerate(zip(bins[:-1], bins[1:])):
    count = ((phase_lengths >= lo) & (phase_lengths < hi)).sum()
    print(f"  {labels[i]:<12} {count:>4} cycles")

print(f"\n── Problematic cycles (phase < 100 timesteps) ──")
if issues:
    for idx, length, group in issues:
        print(f"  idx={idx}  length={length}  group={group}")
else:
    print("  None — all cycles have valid phase data")

# ── Decide crop length ────────────────────────────────────────
# We crop all cycles to the same length
# Use 10th percentile to avoid padding most cycles
crop_len = int(np.percentile(phase_lengths, 10))
pct_needing_crop = (phase_lengths > crop_len).mean() * 100
pct_needing_pad  = (phase_lengths < crop_len).mean() * 100

print(f"\n── Crop length decision ──")
print(f"  10th percentile: {crop_len} timesteps "
      f"= {crop_len * 0.005:.1f} seconds")
print(f"  Cycles longer than crop (will be truncated): "
      f"{pct_needing_crop:.1f}%")
print(f"  Cycles shorter than crop (would need padding): "
      f"{pct_needing_pad:.1f}%")
```

```python
# ============================================================
# Cell 12: TCN Input Array and Model Architecture
# ============================================================

from tensorflow import keras
from tensorflow.keras import layers

# ── Crop/pad strategy ────────────────────────────────────────
# Crop to median phase length — preserves full signal for 50%+
# of cycles, truncates only long tails, minimal padding needed
SEQ_LEN = int(np.median(phase_lengths))  # 5331
N_CHANNELS = 2  # Inj1PrsAct + MldCavPrs1Act

print(f"Sequence length: {SEQ_LEN} timesteps = {SEQ_LEN*0.005:.1f} seconds")
print(f"Channels: {N_CHANNELS} (InjPrs + CavPrs)")
print(f"Building input array: ({len(df)}, {SEQ_LEN}, {N_CHANNELS})")

# ── Extract phase-cropped sequences ──────────────────────────
X_tcn = np.zeros((len(df), SEQ_LEN, N_CHANNELS), dtype=np.float32)

for i, (idx, row) in enumerate(df.iterrows()):
    trig_inj = np.array(row['DXP_TrigInj1'], dtype=bool)
    trig_hld = np.array(row['DXP_TrigHld1'], dtype=bool)
    phase_mask = trig_inj | trig_hld

    inj_prs = np.array(row['DXP_Inj1PrsAct'],   dtype=np.float32)
    cav_prs = np.array(row['DXP_MldCavPrs1Act'], dtype=np.float32)

    # Extract phase segment
    inj_seg = inj_prs[phase_mask]
    cav_seg = cav_prs[phase_mask]

    # Crop or pad to SEQ_LEN
    seg_len = len(inj_seg)
    actual_len = min(seg_len, SEQ_LEN)

    X_tcn[i, :actual_len, 0] = inj_seg[:actual_len]
    X_tcn[i, :actual_len, 1] = cav_seg[:actual_len]
    # Remaining timesteps stay zero (padding)

print(f"\nTCN input shape: {X_tcn.shape}")
print(f"Memory: {X_tcn.nbytes / 1e6:.1f} MB")

# ── Normalise per channel ─────────────────────────────────────
# Normalise across all cycles per channel
# Use training set stats in CV — for now compute global for shape check
for ch in range(N_CHANNELS):
    ch_mean = X_tcn[:, :, ch].mean()
    ch_std  = X_tcn[:, :, ch].std()
    X_tcn[:, :, ch] = (X_tcn[:, :, ch] - ch_mean) / (ch_std + 1e-8)
    print(f"Channel {ch} normalised: mean={ch_mean:.2f}, std={ch_std:.2f}")

# ── Target arrays ─────────────────────────────────────────────
y_sink    = y_binary['LBL_SinkMarks'].values.astype(np.float32)
y_sprue   = y_binary['LBL_SprueCircle'].values.astype(np.float32)
y_under   = y_binary['LBL_Underfilled'].values.astype(np.float32)
y_str     = y_streaks.values.astype(np.int32)

print(f"\nTarget arrays ready:")
print(f"  y_sink:  {y_sink.shape}  positives={y_sink.sum():.0f}")
print(f"  y_sprue: {y_sprue.shape}  positives={y_sprue.sum():.0f}")
print(f"  y_under: {y_under.shape}  positives={y_under.sum():.0f}")
print(f"  y_str:   {y_str.shape}   classes={np.unique(y_str)}")

# ── Build TCN model ───────────────────────────────────────────
def build_tcn(seq_len, n_channels, n_streak_classes=4):
    inputs = keras.Input(shape=(seq_len, n_channels), name='pressure_input')

    x = inputs

    # TCN blocks with increasing dilation
    for filters, dilation in [(32,1),(32,2),(64,4),(64,8)]:
        # Residual connection
        residual = x
        x = layers.Conv1D(
            filters=filters,
            kernel_size=8,
            dilation_rate=dilation,
            padding='causal',  # no future leakage
            activation='relu'
        )(x)
        x = layers.Dropout(0.2)(x)
        x = layers.Conv1D(
            filters=filters,
            kernel_size=8,
            dilation_rate=dilation,
            padding='causal',
            activation='relu'
        )(x)
        x = layers.Dropout(0.2)(x)

        # Match dimensions for residual if needed
        if residual.shape[-1] != filters:
            residual = layers.Conv1D(filters, kernel_size=1)(residual)
        x = layers.Add()([x, residual])

    # Global average pooling — collapses timesteps
    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(64, activation='relu')(x)
    x = layers.Dropout(0.3)(x)

    # Four output heads
    out_sink  = layers.Dense(1, activation='sigmoid', name='SinkMarks')(x)
    out_sprue = layers.Dense(1, activation='sigmoid', name='SprueCircle')(x)
    out_under = layers.Dense(1, activation='sigmoid', name='Underfilled')(x)
    out_str   = layers.Dense(n_streak_classes,
                              activation='softmax', name='Streaks')(x)

    model = keras.Model(inputs=inputs,
                        outputs=[out_sink, out_sprue, out_under, out_str])
    return model

model = build_tcn(SEQ_LEN, N_CHANNELS)
model.summary()

print(f"\nModel built successfully.")
print(f"Input:  (batch, {SEQ_LEN}, {N_CHANNELS})")
print(f"Outputs: SinkMarks, SprueCircle, Underfilled (sigmoid) + Streaks (softmax)")
```

```python
# ============================================================
# Cell 13: TCN GroupKFold Training and Evaluation
# ============================================================

from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import GroupKFold
from sklearn.metrics import f1_score
import warnings
warnings.filterwarnings('ignore')

# ── Raw (un-normalised) TCN input ────────────────────────────
# We normalise per fold inside CV loop to prevent leakage
X_tcn_raw = np.zeros((len(df), SEQ_LEN, N_CHANNELS), dtype=np.float32)

for i, (idx, row) in enumerate(df.iterrows()):
    trig_inj  = np.array(row['DXP_TrigInj1'], dtype=bool)
    trig_hld  = np.array(row['DXP_TrigHld1'], dtype=bool)
    phase_mask = trig_inj | trig_hld

    inj_prs = np.array(row['DXP_Inj1PrsAct'],   dtype=np.float32)
    cav_prs = np.array(row['DXP_MldCavPrs1Act'], dtype=np.float32)

    inj_seg = inj_prs[phase_mask]
    cav_seg = cav_prs[phase_mask]

    actual_len = min(len(inj_seg), SEQ_LEN)
    X_tcn_raw[i, :actual_len, 0] = inj_seg[:actual_len]
    X_tcn_raw[i, :actual_len, 1] = cav_seg[:actual_len]

print(f"Raw TCN input ready: {X_tcn_raw.shape}")

# ── Class weights for imbalanced targets ─────────────────────
def get_class_weight(y):
    n_pos = y.sum()
    n_neg = len(y) - n_pos
    return {0: len(y)/(2*n_neg), 1: len(y)/(2*n_pos)}

# ── GroupKFold TCN training ───────────────────────────────────
gkf    = GroupKFold(n_splits=5)
groups = df['EXP_GROUP'].values

# Store per-fold F1 for each target
tcn_results = {
    'LBL_SinkMarks':  [],
    'LBL_SprueCircle': [],
    'LBL_Underfilled': [],
    'LBL_Streaks':    []
}

for fold, (train_idx, test_idx) in enumerate(
        gkf.split(X_tcn_raw, y_sink, groups)):

    print(f"\n── Fold {fold+1}/5 ──────────────────────────────")

    # ── Normalise using training fold stats only ──────────────
    X_tr_raw = X_tcn_raw[train_idx]
    X_te_raw = X_tcn_raw[test_idx]

    X_tr = X_tr_raw.copy()
    X_te = X_te_raw.copy()

    for ch in range(N_CHANNELS):
        ch_mean = X_tr[:, :, ch].mean()
        ch_std  = X_tr[:, :, ch].std() + 1e-8
        X_tr[:, :, ch] = (X_tr[:, :, ch] - ch_mean) / ch_std
        X_te[:, :, ch] = (X_te[:, :, ch] - ch_mean) / ch_std

    # ── Targets ───────────────────────────────────────────────
    y_tr_sink  = y_sink[train_idx]
    y_te_sink  = y_sink[test_idx]
    y_tr_sprue = y_sprue[train_idx]
    y_te_sprue = y_sprue[test_idx]
    y_tr_under = y_under[train_idx]
    y_te_under = y_under[test_idx]
    y_tr_str   = tf.keras.utils.to_categorical(y_str[train_idx], 4)
    y_te_str   = y_str[test_idx]

    # ── Build fresh model for each fold ──────────────────────
    tf.keras.backend.clear_session()
    model = build_tcn(SEQ_LEN, N_CHANNELS)

    # ── Compile ───────────────────────────────────────────────
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss={
            'SinkMarks':   'binary_crossentropy',
            'SprueCircle': 'binary_crossentropy',
            'Underfilled': 'binary_crossentropy',
            'Streaks':     'categorical_crossentropy'
        },
        loss_weights={
            'SinkMarks':   1.0,
            'SprueCircle': 1.0,
            'Underfilled': 1.0,
            'Streaks':     1.0
        }
    )

    # ── Early stopping ────────────────────────────────────────
    early_stop = keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=10,
        restore_best_weights=True,
        verbose=0
    )

    # ── Train ─────────────────────────────────────────────────
    history = model.fit(
        X_tr,
        {
            'SinkMarks':   y_tr_sink,
            'SprueCircle': y_tr_sprue,
            'Underfilled': y_tr_under,
            'Streaks':     y_tr_str
        },
        validation_data=(
            X_te,
            {
                'SinkMarks':   y_te_sink,
                'SprueCircle': y_te_sprue,
                'Underfilled': y_te_under,
                'Streaks':     tf.keras.utils.to_categorical(y_te_str, 4)
            }
        ),
        epochs=100,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )

    epochs_run = len(history.history['loss'])
    print(f"  Stopped at epoch {epochs_run}")

    # ── Predict & evaluate ────────────────────────────────────
    preds = model.predict(X_te, verbose=0)

    # Binary predictions at 0.5 threshold
    p_sink  = (preds[0].squeeze() > 0.5).astype(int)
    p_sprue = (preds[1].squeeze() > 0.5).astype(int)
    p_under = (preds[2].squeeze() > 0.5).astype(int)
    p_str   = np.argmax(preds[3], axis=1)

    f1_sink  = f1_score(y_te_sink,  p_sink,  average='macro', zero_division=0)
    f1_sprue = f1_score(y_te_sprue, p_sprue, average='macro', zero_division=0)
    f1_under = f1_score(y_te_under, p_under, average='macro', zero_division=0)
    f1_str   = f1_score(y_te_str,   p_str,   average='macro', zero_division=0)

    tcn_results['LBL_SinkMarks'].append(f1_sink)
    tcn_results['LBL_SprueCircle'].append(f1_sprue)
    tcn_results['LBL_Underfilled'].append(f1_under)
    tcn_results['LBL_Streaks'].append(f1_str)

    print(f"  SinkMarks   F1={f1_sink:.3f}")
    print(f"  SprueCircle F1={f1_sprue:.3f}")
    print(f"  Underfilled F1={f1_under:.3f}")
    print(f"  Streaks     F1={f1_str:.3f}")

# ── Final summary ─────────────────────────────────────────────
print("\n" + "="*60)
print("TCN RESULTS — GroupKFold(5)")
print("="*60)
print(f"{'Target':<25} {'Mean F1':>8} {'Std':>8} {'Folds'}")
print("-"*60)
for target, scores in tcn_results.items():
    mean = np.mean(scores)
    std  = np.std(scores)
    fold_str = [f'{s:.2f}' for s in scores]
    print(f"{target:<25} {mean:>8.3f} {std:>8.3f}  {fold_str}")

# ── Compare all three approaches ─────────────────────────────
print("\n" + "="*60)
print("FINAL COMPARISON — All approaches")
print("="*60)
print(f"{'Target':<25} {'LGBM Scalar':>12} {'LGBM+DXP':>10} {'TCN':>8}")
print("-"*60)
all_targets = ['LBL_SinkMarks','LBL_SprueCircle','LBL_Underfilled','LBL_Streaks']
for target in all_targets:
    a = results_A[target][0]
    b = results_B[target][0]
    c = np.mean(tcn_results[target])
    best = max(a, b, c)
    marker_a = '★' if a == best else ' '
    marker_b = '★' if b == best else ' '
    marker_c = '★' if c == best else ' '
    print(f"{target:<25} {a:>10.3f}{marker_a} {b:>8.3f}{marker_b} "
          f"{c:>6.3f}{marker_c}")
```

```python
# ============================================================
# Cell 14: Hyperparameter Tuning — Nested GroupKFold CV
# ============================================================

from lightgbm import LGBMClassifier
from sklearn.model_selection import GroupKFold
from sklearn.metrics import f1_score
from itertools import product
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# ── Search space ─────────────────────────────────────────────
param_grid = {
    'num_leaves':        [15, 31, 63],
    'min_child_samples': [5, 10, 20],
    'learning_rate':     [0.01, 0.05, 0.1],
    'reg_lambda':        [0, 0.1, 1.0],
}

# Generate all combinations
all_combos = [
    dict(zip(param_grid.keys(), vals))
    for vals in product(*param_grid.values())
]
print(f"Total parameter combinations: {len(all_combos)}")
print(f"Total inner fits per target:  {len(all_combos) * 3}")

# ── Nested CV function ────────────────────────────────────────
def nested_cv_tune(X_input, y_input, task, target_name):
    """
    Outer GroupKFold(5) for evaluation.
    Inner GroupKFold(3) on training fold for tuning.
    Returns mean outer F1 per param combo + best params.
    """
    outer_gkf = GroupKFold(n_splits=5)
    inner_gkf = GroupKFold(n_splits=3)

    outer_f1s = []
    best_params_per_fold = []

    for outer_fold, (train_idx, test_idx) in enumerate(
            outer_gkf.split(X_input, y_input, groups)):

        X_outer_tr = X_input.iloc[train_idx]
        y_outer_tr = y_input.iloc[train_idx]
        X_outer_te = X_input.iloc[test_idx]
        y_outer_te = y_input.iloc[test_idx]
        groups_inner = groups[train_idx]

        # ── Inner loop: find best params ──────────────────────
        best_inner_f1 = -1
        best_params   = all_combos[0]

        for params in all_combos:
            inner_f1s = []

            for _, (in_tr, in_te) in enumerate(
                    inner_gkf.split(X_outer_tr, y_outer_tr,
                                    groups_inner)):

                X_in_tr = X_outer_tr.iloc[in_tr]
                y_in_tr = y_outer_tr.iloc[in_tr]
                X_in_te = X_outer_tr.iloc[in_te]
                y_in_te = y_outer_tr.iloc[in_te]

                if task == 'binary':
                    clf = LGBMClassifier(
                        n_estimators=300,
                        class_weight='balanced',
                        objective='binary',
                        random_state=42,
                        verbose=-1,
                        **params
                    )
                    avg = 'macro'
                else:
                    clf = LGBMClassifier(
                        n_estimators=300,
                        class_weight='balanced',
                        objective='multiclass',
                        num_class=4,
                        random_state=42,
                        verbose=-1,
                        **params
                    )
                    avg = 'macro'

                clf.fit(X_in_tr, y_in_tr)
                pred = clf.predict(X_in_te)
                f1 = f1_score(y_in_te, pred,
                              average=avg, zero_division=0)
                inner_f1s.append(f1)

            mean_inner = np.mean(inner_f1s)
            if mean_inner > best_inner_f1:
                best_inner_f1 = mean_inner
                best_params   = params

        best_params_per_fold.append(best_params)

        # ── Outer evaluation with best params ─────────────────
        if task == 'binary':
            clf_final = LGBMClassifier(
                n_estimators=300,
                class_weight='balanced',
                objective='binary',
                random_state=42,
                verbose=-1,
                **best_params
            )
        else:
            clf_final = LGBMClassifier(
                n_estimators=300,
                class_weight='balanced',
                objective='multiclass',
                num_class=4,
                random_state=42,
                verbose=-1,
                **best_params
            )

        clf_final.fit(X_outer_tr, y_outer_tr)
        pred_outer = clf_final.predict(X_outer_te)
        f1_outer = f1_score(y_outer_te, pred_outer,
                            average='macro', zero_division=0)
        outer_f1s.append(f1_outer)

        print(f"  Fold {outer_fold+1}: "
              f"best_params={best_params}  "
              f"outer_F1={f1_outer:.3f}")

    mean_f1 = np.mean(outer_f1s)
    std_f1  = np.std(outer_f1s)

    # Most common best params across folds — use for final model
    # Simple vote: pick params that appeared most as best
    from collections import Counter
    param_votes = Counter(
        [str(p) for p in best_params_per_fold]
    )
    most_common_str = param_votes.most_common(1)[0][0]
    # Recover dict from string match
    final_best = next(
        p for p in best_params_per_fold
        if str(p) == most_common_str
    )

    return mean_f1, std_f1, outer_f1s, final_best

# ── Run nested CV for all targets ────────────────────────────
tuning_results = {}
best_params_all = {}

targets_to_tune = {
    'LBL_SinkMarks':   (X, y_binary['LBL_SinkMarks'],  'binary'),
    'LBL_SprueCircle': (X, y_binary['LBL_SprueCircle'],'binary'),
    'LBL_Underfilled': (X, y_binary['LBL_Underfilled'], 'binary'),
    'LBL_Streaks':     (X, y_streaks,                   'multiclass'),
}

for target, (X_input, y_input, task) in targets_to_tune.items():
    print(f"\n{'='*60}")
    print(f"Tuning: {target}")
    print(f"{'='*60}")
    mean_f1, std_f1, folds, best_p = nested_cv_tune(
        X_input, y_input, task, target
    )
    tuning_results[target] = (mean_f1, std_f1, folds)
    best_params_all[target] = best_p
    print(f"\n  ── Result: F1={mean_f1:.3f} ± {std_f1:.3f} ──")
    print(f"  Best params: {best_p}")

# ── Summary: tuned vs untuned ─────────────────────────────────
print(f"\n{'='*60}")
print("TUNING SUMMARY — Nested CV vs Untuned")
print(f"{'='*60}")
print(f"{'Target':<25} {'Untuned F1':>11} {'Tuned F1':>10} "
      f"{'Delta':>8}")
print("-"*58)

untuned = {
    'LBL_SinkMarks':   results_A['LBL_SinkMarks'][0],
    'LBL_SprueCircle': results_A['LBL_SprueCircle'][0],
    'LBL_Underfilled': results_A['LBL_Underfilled'][0],
    'LBL_Streaks':     results_A['LBL_Streaks'][0],
}

for target in targets_to_tune:
    u = untuned[target]
    t = tuning_results[target][0]
    delta = t - u
    direction = '↑' if delta > 0.01 else ('↓' if delta < -0.01 else '→')
    print(f"  {target:<23} {u:>11.3f} {t:>10.3f} "
          f"{delta:>+8.3f} {direction}")

print(f"\n── Best hyperparameters per target ──")
for target, params in best_params_all.items():
    print(f"  {target:<25} {params}")
```

## Evaluation

**CRISP-DM Phase 5.** With tuned models trained, we now interrogate *why* they make the predictions they do, and use that interrogation to test an assumption made back in Data Preparation.

— Run SHAP TreeExplainer on the final tuned models for all four original targets (SinkMarks, SprueCircle, Underfilled, and the original 4-class ordinal Streaks), and inspect the top contributing features per target.

— Break down the Streaks SHAP values per severity level individually. This is the diagnostic step that exposes the problem: each level is driven by a completely different, sometimes opposing, set of features — contradicting the original assumption that Streaks is a single ordinal severity scale.

— Acting on the Cell 17 finding, replace the single ordinal Streaks target with three independent binary classifiers, retrain, and re-run SHAP. All three levels improve, which is itself additional evidence that the correction was right, not just theoretically motivated.

— Run the secondary 70/30 random-split evaluation as a sanity check, and assemble the complete results summary (GroupKFold vs. random split gap, per-target F1, honest limitations) used throughout the report and presentation.

```python
# ============================================================
# Cell 15: SHAP Analysis on Tuned Final Models
# ============================================================

import shap
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# ── Final model configs ───────────────────────────────────────
final_configs = {
    'LBL_SinkMarks': {
        'X': X, 'y': y_binary['LBL_SinkMarks'],
        'task': 'binary',
        'params': best_params_all['LBL_SinkMarks']
    },
    'LBL_SprueCircle': {
        'X': X, 'y': y_binary['LBL_SprueCircle'],
        'task': 'binary',
        'params': best_params_all['LBL_SprueCircle']
    },
    'LBL_Underfilled': {
        'X': X, 'y': y_binary['LBL_Underfilled'],
        'task': 'binary',
        'params': best_params_all['LBL_Underfilled']
    },
    'LBL_Streaks': {
        'X': X, 'y': y_streaks,
        'task': 'multiclass',
        'params': best_params_all['LBL_Streaks']
    },
}

final_models  = {}
shap_values_dict = {}

for target, cfg in final_configs.items():
    print(f"Training final model: {target}")

    if cfg['task'] == 'binary':
        clf = LGBMClassifier(
            n_estimators=300,
            class_weight='balanced',
            objective='binary',
            random_state=42,
            verbose=-1,
            **cfg['params']
        )
    else:
        clf = LGBMClassifier(
            n_estimators=300,
            class_weight='balanced',
            objective='multiclass',
            num_class=4,
            random_state=42,
            verbose=-1,
            **cfg['params']
        )

    clf.fit(cfg['X'], cfg['y'])
    final_models[target] = clf

    explainer = shap.TreeExplainer(clf)
    shap_vals = explainer.shap_values(cfg['X'])
    shap_values_dict[target] = shap_vals
    print(f"  Done.")

# ── Numeric top-5 SHAP per binary defect ─────────────────────
print("\n── Top 5 SHAP features per defect ──")
print("Positive mean SHAP = feature increases defect risk\n")

binary_targets = ['LBL_SinkMarks', 'LBL_SprueCircle', 'LBL_Underfilled']

for target in binary_targets:
    sv = shap_values_dict[target]
    # Binary LGBM returns list [class0, class1] — we want class1
    if isinstance(sv, list):
        sv = sv[1]

    mean_abs = np.abs(sv).mean(axis=0)
    top5_idx = np.argsort(mean_abs)[-5:][::-1]

    print(f"{target.replace('LBL_','')}:")
    for idx in top5_idx:
        fname    = X.columns[idx]
        mean_val = sv[:, idx].mean()
        direction = '↑ risk' if mean_val > 0 else '↓ risk'
        print(f"  {fname:<42} mean SHAP={mean_val:+.4f}  {direction}")
    print()

# ── Streaks top-5 per class ───────────────────────────────────
print("LBL_Streaks (top 5 per severity level):")
sv_str = shap_values_dict['LBL_Streaks']
streak_labels = ['No Streaks','Level 1','Level 2','Level 3']

if isinstance(sv_str, list):
    for cls_idx, cls_name in enumerate(streak_labels):
        sv_cls = sv_str[cls_idx]
        mean_abs = np.abs(sv_cls).mean(axis=0)
        top5_idx = np.argsort(mean_abs)[-5:][::-1]
        print(f"\n  {cls_name}:")
        for idx in top5_idx:
            fname    = X.columns[idx]
            mean_val = sv_cls[:, idx].mean()
            direction = '↑' if mean_val > 0 else '↓'
            print(f"    {fname:<42} {mean_val:+.4f} {direction}")

# ── SHAP beeswarm plots — one per binary target ───────────────
fig, axes = plt.subplots(1, 3, figsize=(21, 8))

for ax, target in zip(axes, binary_targets):
    plt.sca(ax)
    sv = shap_values_dict[target]
    if isinstance(sv, list):
        sv = sv[1]

    mean_abs = np.abs(sv).mean(axis=0)
    top15_idx   = np.argsort(mean_abs)[-15:]
    top15_names = X.columns[top15_idx].tolist()

    shap.summary_plot(
        sv[:, top15_idx],
        X.iloc[:, top15_idx],
        feature_names=top15_names,
        show=False,
        plot_size=None,
        max_display=15
    )
    ax.set_title(target.replace('LBL_',''),
                 fontsize=12, fontweight='bold')

plt.suptitle(
    'SHAP Feature Importance — Top 15 per Defect Type\n'
    'Colour = feature value (red=high, blue=low)  '
    'X-axis = impact on defect prediction',
    fontsize=11, y=1.02
)
plt.tight_layout()
plt.savefig('shap_beeswarm.png', dpi=130, bbox_inches='tight')
plt.show()
print("\nSaved: shap_beeswarm.png")

# ── SHAP bar plot — mean absolute importance ──────────────────
fig, axes = plt.subplots(1, 4, figsize=(24, 7))

all_targets = ['LBL_SinkMarks', 'LBL_SprueCircle',
               'LBL_Underfilled', 'LBL_Streaks']

for ax, target in zip(axes, all_targets):
    sv = shap_values_dict[target]

    if target == 'LBL_Streaks':
        # Multiclass SHAP stored as ndarray (564, 63, 4)
        # Take mean absolute value across all 4 classes
        if isinstance(sv, np.ndarray) and sv.ndim == 3:
            sv_combined = np.abs(sv).mean(axis=-1)  # (564, 63)
        elif isinstance(sv, list):
            sv_combined = np.mean([np.abs(s) for s in sv], axis=0)
        else:
            sv_combined = np.abs(sv)
    else:
        # Binary: list [class0, class1] — use class1 (defect)
        if isinstance(sv, list):
            sv_combined = np.abs(sv[1])
        else:
            sv_combined = np.abs(sv)

    mean_abs    = sv_combined.mean(axis=0)       # (63,)
    top10_idx   = np.argsort(mean_abs)[-10:]
    top10_names = [X.columns[i] for i in top10_idx]
    top10_vals  = mean_abs[top10_idx]            # guaranteed 1D

    ax.barh(range(10), top10_vals, color='steelblue', alpha=0.8)
    ax.set_yticks(range(10))
    ax.set_yticklabels(top10_names, fontsize=8)
    ax.set_xlabel('Mean |SHAP|', fontsize=9)
    ax.set_title(target.replace('LBL_', ''),
                 fontsize=11, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)

plt.suptitle('Mean Absolute SHAP — Top 10 Features per Target',
             fontsize=12)
plt.tight_layout()
plt.savefig('shap_bar.png', dpi=130, bbox_inches='tight')
plt.show()
print("Saved: shap_bar.png")
```

```python
# ============================================================
# Cell 20: Extension — Material-Specific Models PP vs ABS
# EXTENSION: test if material-specific models outperform mixed model.
# Hypothesis: PP and ABS have different rheology and failure modes.
# Finding: mixed model wins or ties on all evaluable targets.
# Reason: LightGBM tree splits already learn material-conditional
# patterns; splitting halves data below reliable evaluation threshold.
# Conclusion: material-specific modelling needs 500+ cycles per material.
# ============================================================

from sklearn.metrics import f1_score
import warnings
warnings.filterwarnings('ignore')

print("── Dataset split by material ──")
pp_idx  = df[df['MET_MaterialName'] == 'PP'].index
abs_idx = df[df['MET_MaterialName'] == 'ABS'].index

X_pp  = X.loc[pp_idx]
X_abs = X.loc[abs_idx]
groups_pp  = df.loc[pp_idx,  'EXP_GROUP'].values
groups_abs = df.loc[abs_idx, 'EXP_GROUP'].values

print(f"  PP:  {len(pp_idx)} parts  "
      f"{len(np.unique(groups_pp))} experiment groups")
print(f"  ABS: {len(abs_idx)} parts  "
      f"{len(np.unique(groups_abs))} experiment groups")

# ── Targets per material ──────────────────────────────────────
targets_mat = {
    'SinkMarks':    (y_binary['LBL_SinkMarks'],
                     best_params_all['LBL_SinkMarks']),
    'SprueCircle':  (y_binary['LBL_SprueCircle'],
                     best_params_all['LBL_SprueCircle']),
    'Underfilled':  (y_binary['LBL_Underfilled'],
                     best_params_all['LBL_Underfilled']),
    'StreaksLevel1': (y_streak1,
                     {'num_leaves':15,'min_child_samples':5,
                      'learning_rate':0.01,'reg_lambda':0.1}),
    'StreaksLevel2': (y_streak2,
                     {'num_leaves':15,'min_child_samples':5,
                      'learning_rate':0.01,'reg_lambda':0.1}),
    'StreaksLevel3': (y_streak3,
                     {'num_leaves':15,'min_child_samples':5,
                      'learning_rate':0.01,'reg_lambda':0.1}),
}

# ── Run GroupKFold per material ───────────────────────────────
def run_material_cv(X_mat, y_full, groups_mat, params,
                    n_splits=5):
    """GroupKFold on a material-specific subset."""
    y_mat = y_full.loc[X_mat.index]

    # Need at least n_splits groups
    n_groups = len(np.unique(groups_mat))
    splits   = min(n_splits, n_groups // 2)
    if splits < 2:
        return None, None, []

    gkf = GroupKFold(n_splits=splits)
    fold_f1s = []

    for train_idx, test_idx in gkf.split(
            X_mat, y_mat, groups_mat):

        X_tr = X_mat.iloc[train_idx]
        y_tr = y_mat.iloc[train_idx]
        X_te = X_mat.iloc[test_idx]
        y_te = y_mat.iloc[test_idx]

        # Skip fold if test has no positives
        if y_te.sum() == 0:
            continue

        clf = LGBMClassifier(
            n_estimators=300,
            class_weight='balanced',
            objective='binary',
            random_state=42,
            verbose=-1,
            **params
        )
        clf.fit(X_tr, y_tr)
        pred = clf.predict(X_te)
        f1   = f1_score(y_te, pred,
                        average='macro', zero_division=0)
        fold_f1s.append(f1)

    if not fold_f1s:
        return None, None, []

    return np.mean(fold_f1s), np.std(fold_f1s), fold_f1s

# ── Run for both materials ────────────────────────────────────
mat_results = {'PP': {}, 'ABS': {}}

for mat_name, X_mat, groups_mat in [
    ('PP',  X_pp,  groups_pp),
    ('ABS', X_abs, groups_abs)
]:
    print(f"\n── {mat_name} models ──")
    for target, (y_full, params) in targets_mat.items():
        mean_f1, std_f1, folds = run_material_cv(
            X_mat, y_full, groups_mat, params
        )
        mat_results[mat_name][target] = (mean_f1, std_f1)

        pos_count = y_full.loc[X_mat.index].sum()
        if mean_f1 is None:
            print(f"  {target:<18} SKIPPED "
                  f"(insufficient positive samples: {pos_count})")
        else:
            fold_str = [f'{f:.2f}' for f in folds]
            print(f"  {target:<18} F1={mean_f1:.3f} "
                  f"± {std_f1:.3f}  "
                  f"pos={pos_count}  folds={fold_str}")

# ── Comparison table ──────────────────────────────────────────
print(f"\n{'='*72}")
print("MATERIAL-SPECIFIC vs MIXED MODEL COMPARISON")
print(f"{'='*72}")
print(f"{'Target':<18} {'Mixed':>8} {'PP only':>10} "
      f"{'ABS only':>10} {'PP delta':>10} {'ABS delta':>10}")
print("-"*72)

mixed_f1 = {
    'SinkMarks':    0.915,
    'SprueCircle':  0.848,
    'Underfilled':  0.742,
    'StreaksLevel1': 0.827,
    'StreaksLevel2': 0.786,
    'StreaksLevel3': 0.887,
}

for target in targets_mat:
    mixed  = mixed_f1[target]
    pp_f1  = mat_results['PP'][target][0]
    abs_f1 = mat_results['ABS'][target][0]

    pp_str  = f"{pp_f1:.3f}"  if pp_f1  else "N/A"
    abs_str = f"{abs_f1:.3f}" if abs_f1 else "N/A"
    pp_d    = f"{pp_f1-mixed:+.3f}"  if pp_f1  else "N/A"
    abs_d   = f"{abs_f1-mixed:+.3f}" if abs_f1 else "N/A"

    print(f"  {target:<18} {mixed:>8.3f} {pp_str:>10} "
          f"{abs_str:>10} {pp_d:>10} {abs_d:>10}")

print(f"\n── Interpretation ──")
print("  Positive delta = material-specific model beats mixed model")
print("  Negative delta = mixed model generalises better")
print("  N/A = insufficient positive samples for reliable evaluation")
```

```python
# ============================================================
# Cell 21: NOK Derivation from Individual Defect Predictions
# ============================================================
# LBL_NOK is defined in the dataset as:
# NOK = 1 if SinkMarks=1 OR Underfilled=1, else 0
# We derive it from our individual model predictions
# rather than training a separate model — it's not independent
# ============================================================

from sklearn.model_selection import GroupKFold
from sklearn.metrics import f1_score, classification_report
import numpy as np

print("── NOK Definition (from data description) ──")
print("  NOK = 1 if SinkMarks=1 OR Underfilled=1")
print("  All other defects (SprueCircle, Streaks) do not")
print("  trigger NOK in the original labelling scheme\n")

# ── Ground truth NOK ─────────────────────────────────────────
y_nok_true = df['LBL_NOK'].fillna(0).astype(int)
print(f"── Ground truth NOK distribution ──")
print(f"  NOK parts:  {y_nok_true.sum()} / {len(y_nok_true)}"
      f"  ({100*y_nok_true.mean():.1f}%)")
print(f"  OK parts:   {(y_nok_true==0).sum()} / {len(y_nok_true)}"
      f"  ({100*(y_nok_true==0).mean():.1f}%)\n")

# ── Derive NOK via GroupKFold ─────────────────────────────────
# Train SinkMarks and Underfilled models per fold
# Combine predictions with OR logic
gkf    = GroupKFold(n_splits=5)
groups = df['EXP_GROUP'].values

fold_f1s_nok  = []
fold_f1s_sink = []
fold_f1s_und  = []

for fold, (train_idx, test_idx) in enumerate(
        gkf.split(X, y_nok_true, groups)):

    X_tr = X.iloc[train_idx]
    X_te = X.iloc[test_idx]

    # SinkMarks model
    clf_sink = LGBMClassifier(
        n_estimators=300, class_weight='balanced',
        objective='binary', random_state=42, verbose=-1,
        **best_params_all['LBL_SinkMarks']
    )
    clf_sink.fit(X_tr, y_binary['LBL_SinkMarks'].iloc[train_idx])
    pred_sink = clf_sink.predict(X_te)

    # Underfilled model
    clf_und = LGBMClassifier(
        n_estimators=300, class_weight='balanced',
        objective='binary', random_state=42, verbose=-1,
        **best_params_all['LBL_Underfilled']
    )
    clf_und.fit(X_tr, y_binary['LBL_Underfilled'].iloc[train_idx])
    pred_und = clf_und.predict(X_te)

    # Derive NOK = SinkMarks OR Underfilled
    pred_nok = ((pred_sink == 1) | (pred_und == 1)).astype(int)
    y_te_nok = y_nok_true.iloc[test_idx]

    f1_nok  = f1_score(y_te_nok, pred_nok,
                       average='macro', zero_division=0)
    f1_sink = f1_score(
        y_binary['LBL_SinkMarks'].iloc[test_idx],
        pred_sink, average='macro', zero_division=0)
    f1_und  = f1_score(
        y_binary['LBL_Underfilled'].iloc[test_idx],
        pred_und, average='macro', zero_division=0)

    fold_f1s_nok.append(f1_nok)
    fold_f1s_sink.append(f1_sink)
    fold_f1s_und.append(f1_und)

    print(f"  Fold {fold+1}: "
          f"SinkMarks={f1_sink:.3f}  "
          f"Underfilled={f1_und:.3f}  "
          f"NOK(derived)={f1_nok:.3f}")

print(f"\n── NOK Derived Performance ──")
print(f"  SinkMarks   F1 = {np.mean(fold_f1s_sink):.3f}"
      f" ± {np.std(fold_f1s_sink):.3f}")
print(f"  Underfilled F1 = {np.mean(fold_f1s_und):.3f}"
      f" ± {np.std(fold_f1s_und):.3f}")
print(f"  NOK derived F1 = {np.mean(fold_f1s_nok):.3f}"
      f" ± {np.std(fold_f1s_nok):.3f}")

print(f"\n── Complete Defect Coverage Summary ──")
print(f"{'Defect':<20} {'Type':<12} {'Model':<22} {'F1':>8}")
print("-"*65)
coverage = [
    ('SinkMarks',    'Binary',   'LightGBM scalar',  0.915),
    ('SprueCircle',  'Binary',   'LightGBM scalar',  0.848),
    ('Underfilled',  'Binary',   'TCN raw DXP',       0.742),
    ('StreaksLevel1', 'Binary',  'LightGBM binary',   0.827),
    ('StreaksLevel2', 'Binary',  'LightGBM binary',   0.786),
    ('StreaksLevel3', 'Binary',  'LightGBM binary',   0.887),
    ('NOK',          'Derived',  'SinkMarks OR Underfilled',
     np.mean(fold_f1s_nok)),
    ('OldGranulate', 'Excluded', 'Contamination event — not modelled', 0),
]
for defect, dtype, model, f1 in coverage:
    f1_str = f"{f1:.3f}" if f1 > 0 else "N/A"
    print(f"  {defect:<18} {dtype:<12} {model:<22} {f1_str:>8}")

print(f"\n── What this means ──")
print("  We predict 6 specific failure types per cycle.")
print("  NOK is derived from SinkMarks + Underfilled predictions")
print("  without a separate model — consistent with the dataset")
print("  definition (LBL_NOK = SinkMarks OR Underfilled).")
print("  OldGranulate excluded: contamination event, not process")
print("  defect, missing for 27% of parts.")
```

### Closing the loop 

The coverage table above is the concrete basis for the inference pipeline described in Section 6.1 of the report: six binary predictions per cycle (SinkMarks, SprueCircle, Underfilled, StreaksLevel1/2/3), with NOK derived rather than separately modelled, and OldGranulate excluded as out of scope. This is the full defect coverage the triage layer routes on — every part passing through inference produces exactly this set of outputs, which then determines whether it passes automatically, is auto-flagged, or goes to human review.
