## Comparative Analysis of Dense and Convolutional Neural Networks for Metal Nut Defect Classification

## Project Summary

This project develops and evaluates deep learning models for the automatic classification of metal nuts as either **Good** or **Defective** based on image data. The dataset consists of images containing various defect types, including bent, colour variation, flipped orientation, and surface scratches.

Two neural network architectures are implemented and compared:

1. **Multi-Layer Perceptron (MLP)** – a dense neural network using flattened image data.
2. **Convolutional Neural Network (CNN)** – a deep learning model designed to extract spatial features from images.

### Project Workflow

- Image loading and preprocessing
- Image resizing and normalization
- Data augmentation to improve model generalisation
- Train, validation, and test dataset splitting
- Handling class imbalance using class weights
- Development and training of MLP and CNN models
- Hyperparameter tuning and performance optimisation
- Model evaluation and comparison

### Evaluation Metrics

The models are assessed using the following performance metrics:

- Accuracy
- Precision
- Recall
- F1-Score
- ROC-AUC Score
- Confusion Matrix

### Objective

The primary objective of this project is to investigate the effectiveness of deep learning techniques for automated visual quality inspection and to determine whether a Convolutional Neural Network (CNN) can outperform a traditional Dense Neural Network (MLP) in detecting defective metal nuts.

```python

# ============================================================
# Imports & reproducibility seed
# (same as existing notebook — kept identical so it merges cleanly)
# ============================================================
import os
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, confusion_matrix,
    ConfusionMatrixDisplay, roc_auc_score,
    precision_recall_curve, f1_score
)
from sklearn.utils.class_weight import compute_class_weight

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)
print("GPU/MPS devices available:", tf.config.list_physical_devices('GPU'))
```

```python
# ============================================================
# Load images and assign labels per the brief's rule
#   ok      (label 0) = train/good  +  test/good
#   not ok  (label 1) = test/bent + test/color + test/flip + test/scratch
# ============================================================

BASE_PATH = Path("metal_nut")

OK_DIRS = [
    BASE_PATH / "train" / "good",
    BASE_PATH / "test"  / "good",
]

NOT_OK_DIRS = [
    BASE_PATH / "test" / "bent",
    BASE_PATH / "test" / "color",
    BASE_PATH / "test" / "flip",
    BASE_PATH / "test" / "scratch",
]

all_paths = []
all_labels = []

for d in OK_DIRS:
    for f in sorted(d.glob("*.png")):
        all_paths.append(f)
        all_labels.append(0)   # 0 = ok

for d in NOT_OK_DIRS:
    for f in sorted(d.glob("*.png")):
        all_paths.append(f)
        all_labels.append(1)   # 1 = not ok

all_labels = np.array(all_labels)

print(f"Total images : {len(all_paths)}")
print(f"  ok (0)     : {np.sum(all_labels == 0)}")
print(f"  not ok (1) : {np.sum(all_labels == 1)}")
print(f"  Class ratio (ok:not_ok) : {np.sum(all_labels==0)/np.sum(all_labels==1):.2f} : 1")
```

```python
# ============================================================
# Load, resize, and normalise all images into NumPy arrays
# ============================================================

IMG_SIZE = 128   # 128x128 — good balance of detail vs speed on M2 GPU

def load_images(paths, img_size):
    """Open each image, convert to RGB, resize, scale to [0,1]."""
    images = []
    for p in paths:
        img = Image.open(p).convert("RGB")
        img = img.resize((img_size, img_size))
        arr = np.array(img, dtype=np.float32) / 255.0
        images.append(arr)
    return np.array(images)

X = load_images(all_paths, IMG_SIZE)
y = all_labels

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"Pixel min/max: {X.min():.3f} / {X.max():.3f}")
```

```python
# ============================================================
# Stratified train / validation / test split + class weights
# ============================================================

# Step 1: hold out 15% as the final test set (never touched until the end)
X_trainval, X_test, y_trainval, y_test = train_test_split(
    X, y,
    test_size=0.15,
    stratify=y,
    random_state=SEED
)

# Step 2: split remaining into train (~70% of total) and val (~15% of total)
X_train, X_val, y_train, y_val = train_test_split(
    X_trainval, y_trainval,
    test_size=0.18,
    stratify=y_trainval,
    random_state=SEED
)

print(f"Train samples      : {len(X_train)}  (ok={np.sum(y_train==0)}, not_ok={np.sum(y_train==1)})")
print(f"Validation samples : {len(X_val)}  (ok={np.sum(y_val==0)}, not_ok={np.sum(y_val==1)})")
print(f"Test samples       : {len(X_test)}  (ok={np.sum(y_test==0)}, not_ok={np.sum(y_test==1)})")

# Class weights — penalise mistakes on the minority class ("not ok") more heavily
cw_values = compute_class_weight("balanced", classes=np.array([0, 1]), y=y_train)
class_weight = {0: cw_values[0], 1: cw_values[1]}
print(f"\nBalanced class weights: {class_weight}")
```

```python
# ============================================================
# Data augmentation pipeline
# Applied only during training (inside the model, training=True),
# so validation/test images are never artificially altered.
# ============================================================

augmentation_pipeline = keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),   # nuts can appear in any orientation
    layers.RandomRotation(0.15),                     # ±15% of 360° ≈ ±54°
    layers.RandomZoom(0.10),                         # ±10% zoom — slight scale variation
    layers.RandomContrast(0.10),                     # ±10% contrast — lighting variation
], name="augmentation")

# Sanity check — visualise augmented versions of one sample image
sample_img = X_train[0:1]
fig, axes = plt.subplots(1, 6, figsize=(14, 2.5))
axes[0].imshow(sample_img[0])
axes[0].set_title("Original")
axes[0].axis("off")
for i in range(1, 6):
    aug = augmentation_pipeline(sample_img, training=True)[0].numpy()
    axes[i].imshow(aug)
    axes[i].set_title(f"Aug {i}")
    axes[i].axis("off")
plt.suptitle("Augmentation examples", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Reusable helper: plot training history
# ============================================================

def plot_history(history, title="Training history"):
    """Plot accuracy and loss curves (train vs val) from a Keras history object."""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle(title, fontsize=13, fontweight="bold")

    ax1.plot(history.history["accuracy"], label="Train acc", lw=2)
    ax1.plot(history.history["val_accuracy"], label="Val acc", lw=2, ls="--")
    ax1.set_xlabel("Epoch")
    ax1.set_ylabel("Accuracy")
    ax1.legend()
    ax1.grid(alpha=0.3)

    ax2.plot(history.history["loss"], label="Train loss", lw=2)
    ax2.plot(history.history["val_loss"], label="Val loss", lw=2, ls="--")
    ax2.set_xlabel("Epoch")
    ax2.set_ylabel("Loss")
    ax2.legend()
    ax2.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()


def evaluate_model(model, X_test, y_test, model_name="Model"):
    """Evaluate a trained model on the test set and print/return metrics."""
    y_prob = model.predict(X_test, verbose=0).flatten()
    y_pred = (y_prob > 0.5).astype(int)

    print("\n" + "=" * 50)
    print(f"Results for: {model_name}")
    print("=" * 50)
    report = classification_report(y_test, y_pred, target_names=["ok", "not ok"])
    print(report)

    auc = roc_auc_score(y_test, y_prob)
    print(f"AUC-ROC: {auc:.4f}")

    cm = confusion_matrix(y_test, y_pred)
    fig, ax = plt.subplots(figsize=(4, 3))
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["ok", "not ok"])
    disp.plot(ax=ax, cmap="Blues")
    ax.set_title(f"Confusion Matrix — {model_name}")
    plt.tight_layout()
    plt.show()

    report_dict = classification_report(y_test, y_pred, target_names=["ok", "not ok"], output_dict=True)
    return {
        "Model": model_name,
        "Accuracy": report_dict["accuracy"],
        "Precision (not ok)": report_dict["not ok"]["precision"],
        "Recall (not ok)": report_dict["not ok"]["recall"],
        "F1 (not ok)": report_dict["not ok"]["f1-score"],
        "AUC-ROC": auc,
        "y_prob": y_prob,
        "y_pred": y_pred,
    }
```

```python
# ============================================================
# Dense (MLP) network — build and train
# ============================================================

def build_mlp(input_shape, dropout_rate=0.4):
    inputs = keras.Input(shape=input_shape, name="image_input")
    x = augmentation_pipeline(inputs)
    x = layers.Flatten()(x)

    x = layers.Dense(256, activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)

    x = layers.Dense(128, activation="relu")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Dropout(dropout_rate)(x)

    x = layers.Dense(64, activation="relu")(x)
    x = layers.Dropout(dropout_rate / 2)(x)

    outputs = layers.Dense(1, activation="sigmoid", name="output")(x)
    return keras.Model(inputs, outputs, name="MLP")


mlp_model = build_mlp(input_shape=(IMG_SIZE, IMG_SIZE, 3))
mlp_model.summary()

mlp_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=3e-4),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

callbacks_mlp = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.5, patience=7, verbose=1
    ),
]

history_mlp = mlp_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=60,
    batch_size=32,
    class_weight=class_weight,
    callbacks=callbacks_mlp,
    verbose=1
)

print("MLP training complete.")
plot_history(history_mlp, title="MLP — Training History")
metrics_mlp = evaluate_model(mlp_model, X_test, y_test, model_name="MLP")
```

```python
# ============================================================
# Convolutional Neural Network (from scratch)
#
# Architecture:
#   Input (128x128x3)
#   -> Conv block 1: Conv2D(32) -> ReLU -> MaxPool
#   -> Conv block 2: Conv2D(64) -> ReLU -> MaxPool
#   -> Conv block 3: Conv2D(128) -> ReLU -> MaxPool
#   -> Flatten -> Dense(128) -> Dropout -> Dense(1, sigmoid)
#
# Why this design:
#   - 3 conv blocks progressively shrink spatial size while growing
#     feature depth (32->64->128), a standard, proven pattern for
#     small-to-medium image classification tasks.
#   - MaxPooling after each conv reduces computation and adds a small
#     amount of translation invariance.
#   - Kept deliberately small (not deeper/wider) given only 232
#     training images — larger CNNs would likely overfit, similar to
#     what we saw with the oversized MLP.
# ============================================================

def build_cnn(input_shape, dropout_rate=0.3):
    inputs = keras.Input(shape=input_shape, name="image_input")

    x = augmentation_pipeline(inputs)

    x = layers.Conv2D(32, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(64, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(128, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Flatten()(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(dropout_rate)(x)

    outputs = layers.Dense(1, activation="sigmoid", name="output")(x)
    return keras.Model(inputs, outputs, name="CNN_from_scratch")


cnn_model = build_cnn(input_shape=(IMG_SIZE, IMG_SIZE, 3))
cnn_model.summary()

cnn_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

callbacks_cnn = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=12, restore_best_weights=True, verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.5, patience=6, verbose=1
    ),
]

history_cnn = cnn_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=40,
    batch_size=32,
    class_weight=class_weight,
    callbacks=callbacks_cnn,
    verbose=1
)

print("CNN (from scratch) training complete.")

plot_history(history_cnn, title="CNN (from scratch) — Training History")
metrics_cnn = evaluate_model(cnn_model, X_test, y_test, model_name="CNN (from scratch)")
```

```python
# ============================================================
# Hyperparameter experiment: Class Weight Strength
# Contributor: Sravanthi Munagapati
# Fix applied: build a fresh augmentation layer INSIDE build_cnn_v2
# for each run (not reusing the shared global augmentation_pipeline),
# to prevent any state leakage between successive training runs in
# the same loop/session.
# ============================================================

def build_cnn_v2(input_shape, dropout_rate=0.3):
    """Same architecture as build_cnn, but with its own fresh augmentation layer."""
    inputs = keras.Input(shape=input_shape, name="image_input")

    local_aug = keras.Sequential([
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(0.15),
        layers.RandomZoom(0.10),
        layers.RandomContrast(0.10),
    ])
    x = local_aug(inputs)

    x = layers.Conv2D(32, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(64, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(128, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Flatten()(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(dropout_rate)(x)

    outputs = layers.Dense(1, activation="sigmoid", name="output")(x)
    return keras.Model(inputs, outputs, name="CNN_from_scratch_v2")


PARAM_NAME = "class_weight_multiplier (not_ok class)"
PARAM_VALUES = [1, 2, 4]
base_weight_not_ok = class_weight[1]

hp_results = {}

for mult in PARAM_VALUES:
    print(f"\n--- {PARAM_NAME} = {mult}x  (not_ok weight = {base_weight_not_ok * mult:.3f}) ---")

    tf.keras.backend.clear_session()   # fully clear graph/state between runs
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    random.seed(SEED)

    scaled_class_weight = {0: class_weight[0], 1: base_weight_not_ok * mult}

    model = build_cnn_v2(input_shape=(IMG_SIZE, IMG_SIZE, 3), dropout_rate=0.3)
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=0
    )

    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=40,
        batch_size=32,
        class_weight=scaled_class_weight,
        callbacks=[es],
        verbose=0
    )

    y_val_prob = model.predict(X_val, verbose=0).flatten()
    y_val_pred = (y_val_prob > 0.5).astype(int)
    report = classification_report(y_val, y_val_pred, target_names=["ok", "not ok"],
                                    output_dict=True, zero_division=0)

    hp_results[mult] = {
        "val_accuracy": report["accuracy"],
        "not_ok_precision": report["not ok"]["precision"],
        "not_ok_recall": report["not ok"]["recall"],
        "not_ok_f1": report["not ok"]["f1-score"],
        "epochs_run": len(hist.history["val_accuracy"]),
    }

    print(f"  Val accuracy: {report['accuracy']:.4f}  |  "
          f"not_ok precision: {report['not ok']['precision']:.4f}  |  "
          f"not_ok recall: {report['not ok']['recall']:.4f}  |  "
          f"not_ok F1: {report['not ok']['f1-score']:.4f}  "
          f"(stopped at epoch {len(hist.history['val_accuracy'])})")

# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 5))
x_labels = [str(v) for v in PARAM_VALUES]
ax.plot(x_labels, [hp_results[v]["not_ok_recall"] for v in PARAM_VALUES],
        marker="o", lw=2, label="not_ok recall", color="crimson")
ax.plot(x_labels, [hp_results[v]["not_ok_precision"] for v in PARAM_VALUES],
        marker="s", lw=2, label="not_ok precision", color="steelblue")
ax.plot(x_labels, [hp_results[v]["val_accuracy"] for v in PARAM_VALUES],
        marker="^", lw=2, label="overall val accuracy", color="gray", linestyle="--")
ax.set_xlabel("Class weight multiplier (applied to 'not ok' class)")
ax.set_ylabel("Score")
ax.set_title("Effect of Class Weight Strength on CNN Performance")
ax.set_ylim(0, 1.05)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nSummary table:")
summary_df = pd.DataFrame(hp_results).T
summary_df.index.name = "weight_multiplier"
print(summary_df.to_string())
```

```python
# ============================================================
# Hyperparameter experiment: Learning Rate
#
# Contributor: Ravi Biswajeet
#
# Why this hyperparameter:
#   Learning rate controls how large a step the optimizer takes
#   when updating weights after each batch. Too high, and training
#   can overshoot and never settle (unstable loss, like we saw with
#   the MLP). Too low, and training is painfully slow to converge,
#   potentially never reaching a good solution within our epoch
#   budget. This experiment finds a reasonable operating range for
#   our from-scratch CNN.
#
# PARAM_VALUES: 1e-4 (conservative), 1e-3 (our current default),
#               1e-2 (aggressive)
# ============================================================

PARAM_NAME = "learning_rate"
PARAM_VALUES = [1e-4, 1e-3, 1e-2]

hp_results_lr = {}

for lr in PARAM_VALUES:
    print(f"\n--- {PARAM_NAME} = {lr} ---")

    tf.keras.backend.clear_session()
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    random.seed(SEED)

    model = build_cnn_v2(input_shape=(IMG_SIZE, IMG_SIZE, 3), dropout_rate=0.3)
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=lr),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=0
    )

    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=40,
        batch_size=32,
        class_weight=class_weight,
        callbacks=[es],
        verbose=0
    )

    y_val_prob = model.predict(X_val, verbose=0).flatten()
    y_val_pred = (y_val_prob > 0.5).astype(int)
    report = classification_report(y_val, y_val_pred, target_names=["ok", "not ok"],
                                    output_dict=True, zero_division=0)

    hp_results_lr[lr] = {
        "val_accuracy": report["accuracy"],
        "not_ok_precision": report["not ok"]["precision"],
        "not_ok_recall": report["not ok"]["recall"],
        "not_ok_f1": report["not ok"]["f1-score"],
        "epochs_run": len(hist.history["val_accuracy"]),
    }

    print(f"  Val accuracy: {report['accuracy']:.4f}  |  "
          f"not_ok precision: {report['not ok']['precision']:.4f}  |  "
          f"not_ok recall: {report['not ok']['recall']:.4f}  |  "
          f"not_ok F1: {report['not ok']['f1-score']:.4f}  "
          f"(stopped at epoch {len(hist.history['val_accuracy'])})")

# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 5))
x_labels = [f"{v:.0e}" for v in PARAM_VALUES]
ax.plot(x_labels, [hp_results_lr[v]["not_ok_recall"] for v in PARAM_VALUES],
        marker="o", lw=2, label="not_ok recall", color="crimson")
ax.plot(x_labels, [hp_results_lr[v]["not_ok_precision"] for v in PARAM_VALUES],
        marker="s", lw=2, label="not_ok precision", color="steelblue")
ax.plot(x_labels, [hp_results_lr[v]["val_accuracy"] for v in PARAM_VALUES],
        marker="^", lw=2, label="overall val accuracy", color="gray", linestyle="--")
ax.set_xlabel("Learning rate")
ax.set_ylabel("Score")
ax.set_title("Effect of Learning Rate on CNN Performance")
ax.set_ylim(0, 1.05)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nSummary table:")
summary_df_lr = pd.DataFrame(hp_results_lr).T
summary_df_lr.index.name = "learning_rate"
print(summary_df_lr.to_string())
```

```python
# ============================================================
# Hyperparameter experiment: Number of Filters (Conv Block 1)
#
# Contributor: Arya Dinesh Chirakkuni
#
# Why this hyperparameter:
#   The number of filters in the first conv layer controls how many
#   distinct low-level visual patterns (edges, textures, color
#   gradients) the network can detect at the earliest stage. Fewer
#   filters = less capacity to notice different things, but lower
#   risk of overfitting on our small dataset. More filters = richer
#   feature detection, but more parameters to learn from only 232
#   training images. Blocks 2 and 3 are kept fixed (64, 128) so we
#   isolate the effect of this one architectural choice.
#
# PARAM_VALUES: 16 (narrow), 32 (current default), 64 (wide)
# ============================================================

def build_cnn_filters(input_shape, block1_filters, dropout_rate=0.3):
    inputs = keras.Input(shape=input_shape, name="image_input")

    local_aug = keras.Sequential([
        layers.RandomFlip("horizontal_and_vertical"),
        layers.RandomRotation(0.15),
        layers.RandomZoom(0.10),
        layers.RandomContrast(0.10),
    ])
    x = local_aug(inputs)

    x = layers.Conv2D(block1_filters, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(64, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Conv2D(128, (3, 3), activation="relu")(x)
    x = layers.MaxPooling2D()(x)

    x = layers.Flatten()(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(dropout_rate)(x)

    outputs = layers.Dense(1, activation="sigmoid", name="output")(x)
    return keras.Model(inputs, outputs, name=f"CNN_filters_{block1_filters}")


PARAM_NAME = "conv_block1_filters"
PARAM_VALUES = [16, 32, 64]

hp_results_filters = {}

for n_filters in PARAM_VALUES:
    print(f"\n--- {PARAM_NAME} = {n_filters} ---")

    tf.keras.backend.clear_session()
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    random.seed(SEED)

    model = build_cnn_filters(input_shape=(IMG_SIZE, IMG_SIZE, 3), block1_filters=n_filters, dropout_rate=0.3)
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=0
    )

    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=40,
        batch_size=32,
        class_weight=class_weight,
        callbacks=[es],
        verbose=0
    )

    n_params = model.count_params()
    y_val_prob = model.predict(X_val, verbose=0).flatten()
    y_val_pred = (y_val_prob > 0.5).astype(int)
    report = classification_report(y_val, y_val_pred, target_names=["ok", "not ok"],
                                    output_dict=True, zero_division=0)

    hp_results_filters[n_filters] = {
        "val_accuracy": report["accuracy"],
        "not_ok_precision": report["not ok"]["precision"],
        "not_ok_recall": report["not ok"]["recall"],
        "not_ok_f1": report["not ok"]["f1-score"],
        "epochs_run": len(hist.history["val_accuracy"]),
        "total_params": n_params,
    }

    print(f"  Val accuracy: {report['accuracy']:.4f}  |  "
          f"not_ok precision: {report['not ok']['precision']:.4f}  |  "
          f"not_ok recall: {report['not ok']['recall']:.4f}  |  "
          f"not_ok F1: {report['not ok']['f1-score']:.4f}  |  "
          f"params: {n_params:,}  "
          f"(stopped at epoch {len(hist.history['val_accuracy'])})")

# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 5))
x_labels = [str(v) for v in PARAM_VALUES]
ax.plot(x_labels, [hp_results_filters[v]["not_ok_recall"] for v in PARAM_VALUES],
        marker="o", lw=2, label="not_ok recall", color="crimson")
ax.plot(x_labels, [hp_results_filters[v]["not_ok_precision"] for v in PARAM_VALUES],
        marker="s", lw=2, label="not_ok precision", color="steelblue")
ax.plot(x_labels, [hp_results_filters[v]["val_accuracy"] for v in PARAM_VALUES],
        marker="^", lw=2, label="overall val accuracy", color="gray", linestyle="--")
ax.set_xlabel("Number of filters in Conv block 1")
ax.set_ylabel("Score")
ax.set_title("Effect of Conv Block 1 Filter Count on CNN Performance")
ax.set_ylim(0, 1.05)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nSummary table:")
summary_df_filters = pd.DataFrame(hp_results_filters).T
summary_df_filters.index.name = "block1_filters"
print(summary_df_filters.to_string())
```

```python
# ============================================================
# Hyperparameter experiment: Batch Size
#
# Contributor: Hanna Susan Mathew
#
# Why this hyperparameter:
#   Batch size controls how many training images the model looks at
#   before updating its weights once. Smaller batches give noisier
#   but more frequent updates (can help escape poor local solutions,
#   but can also be unstable). Larger batches give smoother, more
#   stable gradient estimates, but fewer updates per epoch and can
#   sometimes generalize worse. With class weighting in play, batch
#   size also affects how many minority-class ("not ok") examples
#   appear in a typical batch — at batch size 16, a batch might
#   contain 0 not_ok examples by chance; at 64, it's more likely to
#   contain several.
#
# PARAM_VALUES: 16 (small), 32 (current default), 64 (large)
# ============================================================

PARAM_NAME = "batch_size"
PARAM_VALUES = [16, 32, 64]

hp_results_batch = {}

for bs in PARAM_VALUES:
    print(f"\n--- {PARAM_NAME} = {bs} ---")

    tf.keras.backend.clear_session()
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    random.seed(SEED)

    model = build_cnn_v2(input_shape=(IMG_SIZE, IMG_SIZE, 3), dropout_rate=0.3)
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    es = keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=0
    )

    hist = model.fit(
        X_train, y_train,
        validation_data=(X_val, y_val),
        epochs=40,
        batch_size=bs,
        class_weight=class_weight,
        callbacks=[es],
        verbose=0
    )

    y_val_prob = model.predict(X_val, verbose=0).flatten()
    y_val_pred = (y_val_prob > 0.5).astype(int)
    report = classification_report(y_val, y_val_pred, target_names=["ok", "not ok"],
                                    output_dict=True, zero_division=0)

    hp_results_batch[bs] = {
        "val_accuracy": report["accuracy"],
        "not_ok_precision": report["not ok"]["precision"],
        "not_ok_recall": report["not ok"]["recall"],
        "not_ok_f1": report["not ok"]["f1-score"],
        "epochs_run": len(hist.history["val_accuracy"]),
    }

    print(f"  Val accuracy: {report['accuracy']:.4f}  |  "
          f"not_ok precision: {report['not ok']['precision']:.4f}  |  "
          f"not_ok recall: {report['not ok']['recall']:.4f}  |  "
          f"not_ok F1: {report['not ok']['f1-score']:.4f}  "
          f"(stopped at epoch {len(hist.history['val_accuracy'])})")

# ---- Plot ----
fig, ax = plt.subplots(figsize=(7, 5))
x_labels = [str(v) for v in PARAM_VALUES]
ax.plot(x_labels, [hp_results_batch[v]["not_ok_recall"] for v in PARAM_VALUES],
        marker="o", lw=2, label="not_ok recall", color="crimson")
ax.plot(x_labels, [hp_results_batch[v]["not_ok_precision"] for v in PARAM_VALUES],
        marker="s", lw=2, label="not_ok precision", color="steelblue")
ax.plot(x_labels, [hp_results_batch[v]["val_accuracy"] for v in PARAM_VALUES],
        marker="^", lw=2, label="overall val accuracy", color="gray", linestyle="--")
ax.set_xlabel("Batch size")
ax.set_ylabel("Score")
ax.set_title("Effect of Batch Size on CNN Performance")
ax.set_ylim(0, 1.05)
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nSummary table:")
summary_df_batch = pd.DataFrame(hp_results_batch).T
summary_df_batch.index.name = "batch_size"
print(summary_df_batch.to_string())
```

```python
# ============================================================
# Combining the best hyperparameter findings
# Contributor: Arya Dinesh Chirakkuni
# Individual sweeps found:
#   - Learning rate 1e-4 outperformed 1e-3 (default) and 1e-2
#   - Batch size 64 outperformed 32 (default) and 16
#   - Class weight 1x (balanced) is optimal — 2x/4x both collapsed
#   - 16 or 32 filters in Conv block 1 (no improvement from 64)
#
# This cell tests the combination of the two strongest individual
# findings (LR=1e-4, batch_size=64) together, to check whether their
# benefits stack, since they were only tested independently so far.
# ============================================================

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

combined_model = build_cnn_v2(input_shape=(IMG_SIZE, IMG_SIZE, 3), dropout_rate=0.3)
combined_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-4),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

es_combined = keras.callbacks.EarlyStopping(
    monitor="val_loss", patience=15, restore_best_weights=True, verbose=1
)

history_combined = combined_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=40,
    batch_size=64,
    class_weight=class_weight,
    callbacks=[es_combined],
    verbose=1
)

y_val_prob_combined = combined_model.predict(X_val, verbose=0).flatten()
y_val_pred_combined = (y_val_prob_combined > 0.5).astype(int)
report_combined = classification_report(y_val, y_val_pred_combined,
                                          target_names=["ok", "not ok"],
                                          output_dict=True, zero_division=0)

print("\n--- Combined (LR=1e-4, batch_size=64) — Validation Results ---")
print(classification_report(y_val, y_val_pred_combined, target_names=["ok", "not ok"], zero_division=0))

print("\nComparison so far (validation set, not_ok metrics):")
comparison = pd.DataFrame({
    "Default (LR=1e-3, bs=32)":      {"recall": 0.2857, "precision": 1.0000, "accuracy": 0.8077},
    "Best LR alone (1e-4, bs=32)":   {"recall": 0.3571, "precision": 1.0000, "accuracy": 0.8269},
    "Best batch alone (LR=1e-3, bs=64)": {"recall": 0.5000, "precision": 1.0000, "accuracy": 0.8654},
    "Combined (LR=1e-4, bs=64)":     {"recall": report_combined["not ok"]["recall"],
                                        "precision": report_combined["not ok"]["precision"],
                                        "accuracy": report_combined["accuracy"]},
}).T
print(comparison.round(4).to_string())
```

```python
# ============================================================
# Finalize best CNN configuration and
# evaluate on the held-out TEST set.
#
# Configuration chosen based on hyperparameter sweeps:
#   - Class weight: balanced / 1x (Cell 10 sweep — 2x/4x both
#     collapsed training entirely)
#   - Learning rate: 1e-3 / default (Cell 16 sweep found 1e-4
#     slightly better in isolation, but combined with batch_size=64
#     it performed WORSE — see Cell 19 — so we keep the default
#     learning rate here, consistent with the known interaction
#     between batch size and learning rate)
#   - Conv block 1 filters: 32 / default (Cell 17 sweep — 16/32
#     performed identically, 64 hurt recall)
#   - Batch size: 64 (Cell 18 sweep — clear winner, nearly doubled
#     recall vs the batch_size=32 default)
# ============================================================

tf.keras.backend.clear_session()
tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

cnn_model_final = build_cnn_v2(input_shape=(IMG_SIZE, IMG_SIZE, 3), dropout_rate=0.3)
cnn_model_final.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

callbacks_final = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss", patience=15, restore_best_weights=True, verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss", factor=0.5, patience=7, verbose=1
    ),
]

history_cnn_final = cnn_model_final.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=40,
    batch_size=64,
    class_weight=class_weight,
    callbacks=callbacks_final,
    verbose=1
)

print("\nFinal CNN training complete (batch_size=64 configuration).")

plot_history(history_cnn_final, title="Final CNN (batch_size=64) — Training History")
metrics_cnn_final = evaluate_model(cnn_model_final, X_test, y_test, model_name="CNN (final, batch_size=64)")
```

```python
# ============================================================
# Threshold tuning
# Contributor: Hanna Susan Mathew
# 
# So far we've used the default decision threshold of 0.5 (if the
# model's predicted probability of "not ok" is above 0.5, we call it
# defective). But 0.5 is just a default — it's not necessarily the
# best cutoff for our specific goal: catching as many real defects
# as possible, since missing a defect is the costliest mistake in
# quality inspection.
#
# Here we sweep the threshold from low to high and see how
# precision and recall trade off, so we can pick a threshold that
# better balances "catch more defects" against "don't cry wolf too
# often" — instead of blindly using 0.5.
#
# IMPORTANT: we use the VALIDATION set to choose the threshold
# (never the test set, to avoid leaking test information into our
# decision), then apply the chosen threshold once to the test set
# at the end for final, honest reporting.
# ============================================================

y_val_prob_final = cnn_model_final.predict(X_val, verbose=0).flatten()

precisions, recalls, thresholds = precision_recall_curve(y_val, y_val_prob_final)

f1_scores = 2 * (precisions * recalls) / (precisions + recalls + 1e-9)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(thresholds, precisions[:-1], label="Precision (not ok)", lw=2, color="steelblue")
ax.plot(thresholds, recalls[:-1], label="Recall (not ok)", lw=2, color="crimson")
ax.plot(thresholds, f1_scores[:-1], label="F1 (not ok)", lw=2, color="green", linestyle="--")
ax.axvline(0.5, color="gray", linestyle=":", label="Default threshold (0.5)")
ax.set_xlabel("Decision threshold")
ax.set_ylabel("Score")
ax.set_title("Precision / Recall / F1 vs Threshold (validation set)")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

best_f1_idx = np.argmax(f1_scores[:-1])
best_f1_threshold = thresholds[best_f1_idx]
print(f"Threshold that maximizes F1 on validation set: {best_f1_threshold:.3f}")
print(f"  -> Precision: {precisions[best_f1_idx]:.3f}, Recall: {recalls[best_f1_idx]:.3f}, F1: {f1_scores[best_f1_idx]:.3f}")

target_recall = 0.70
candidates = [(t, p, r) for t, p, r in zip(thresholds, precisions[:-1], recalls[:-1]) if r >= target_recall]
if candidates:
    best_recall_candidate = max(candidates, key=lambda x: x[1])
    print(f"\nLowest threshold achieving >= {target_recall:.0%} recall: {best_recall_candidate[0]:.3f}")
    print(f"  -> Precision: {best_recall_candidate[1]:.3f}, Recall: {best_recall_candidate[2]:.3f}")
else:
    print(f"\nNo threshold achieves >= {target_recall:.0%} recall on the validation set.")
```

```python
# ============================================================
# Apply the chosen threshold (selected on validation data
# only) to the TEST set for final, honest reporting.
# ============================================================

CHOSEN_THRESHOLD = 0.411   # selected via F1-maximization on the validation set (Cell 12, batch_size=64 model)

y_test_prob = cnn_model_final.predict(X_test, verbose=0).flatten()
y_test_pred_default = (y_test_prob > 0.5).astype(int)
y_test_pred_tuned = (y_test_prob > CHOSEN_THRESHOLD).astype(int)

print("=" * 55)
print("TEST SET — Default threshold (0.5)")
print("=" * 55)
print(classification_report(y_test, y_test_pred_default, target_names=["ok", "not ok"], zero_division=0))

print("=" * 55)
print(f"TEST SET — Tuned threshold ({CHOSEN_THRESHOLD})")
print("=" * 55)
print(classification_report(y_test, y_test_pred_tuned, target_names=["ok", "not ok"], zero_division=0))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, y_pred, title in zip(
    axes,
    [y_test_pred_default, y_test_pred_tuned],
    [f"Default threshold (0.5)", f"Tuned threshold ({CHOSEN_THRESHOLD})"]
):
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["ok", "not ok"])
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(title)
plt.tight_layout()
plt.show()

n_defects = int(np.sum(y_test == 1))
n_caught_default = int(np.sum((y_test == 1) & (y_test_pred_default == 1)))
n_caught_tuned = int(np.sum((y_test == 1) & (y_test_pred_tuned == 1)))
print(f"\nOut of {n_defects} actual defects in the test set:")
print(f"  Default threshold (0.5)  caught {n_caught_default}  (missed {n_defects - n_caught_default})")
print(f"  Tuned threshold ({CHOSEN_THRESHOLD}) caught {n_caught_tuned}  (missed {n_defects - n_caught_tuned})")
```

```python
# ============================================================
# Final model comparison
# 
# Compares MLP vs CNN (default threshold) vs CNN (tuned threshold)
# on the held-out test set, side by side.
# ============================================================

def get_test_metrics(y_true, y_pred, y_prob, model_name):
    report = classification_report(y_true, y_pred, target_names=["ok", "not ok"],
                                     output_dict=True, zero_division=0)
    auc = roc_auc_score(y_true, y_prob)
    return {
        "Model": model_name,
        "Accuracy": report["accuracy"],
        "Precision (not ok)": report["not ok"]["precision"],
        "Recall (not ok)": report["not ok"]["recall"],
        "F1 (not ok)": report["not ok"]["f1-score"],
        "AUC-ROC": auc,
    }

mlp_row = {
    "Model": "MLP",
    "Accuracy": metrics_mlp["Accuracy"],
    "Precision (not ok)": metrics_mlp["Precision (not ok)"],
    "Recall (not ok)": metrics_mlp["Recall (not ok)"],
    "F1 (not ok)": metrics_mlp["F1 (not ok)"],
    "AUC-ROC": metrics_mlp["AUC-ROC"],
}

cnn_default_row = get_test_metrics(y_test, y_test_pred_default, y_test_prob, "CNN (default thr. 0.5)")
cnn_tuned_row = get_test_metrics(y_test, y_test_pred_tuned, y_test_prob, f"CNN (tuned thr. {CHOSEN_THRESHOLD})")

results_table = pd.DataFrame([mlp_row, cnn_default_row, cnn_tuned_row]).set_index("Model")

print("=" * 70)
print("        FINAL MODEL COMPARISON (Test Set)")
print("=" * 70)
print(results_table.round(3).to_string())
print("=" * 70)

ax = results_table.round(3).plot(kind="bar", figsize=(10, 5), rot=15, ylim=(0, 1.05))
plt.title("Model Comparison — Test Set Metrics")
plt.ylabel("Score")
plt.legend(loc="lower right", fontsize=8)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

print("\nKey takeaways:")
print(f"  - MLP:  recall {mlp_row['Recall (not ok)']:.2f} on minority class — unstable training, weakest discriminator")
print(f"  - CNN (default threshold): recall {cnn_default_row['Recall (not ok)']:.2f} — high precision but misses most defects")
print(f"  - CNN (tuned threshold):   recall {cnn_tuned_row['Recall (not ok)']:.2f} — best defect-catching rate, "
      f"with a controlled drop in precision ({cnn_tuned_row['Precision (not ok)']:.2f})")
```

```python
# ============================================================
# Per-defect-type error breakdown
#
# Contributor: Sravanthi Munagapati
# y_test only tells us "ok" vs "not ok" — it doesn't tell us WHICH
# defect type (bent/color/flip/scratch) each "not ok" test image
# actually is. To find that out, we need to track defect type
# alongside the path/label from the very start, then filter down
# to just the test set indices.
#
# Since X/y were built from all_paths/all_labels (in that exact
# order) and split with the SAME random_state/stratify calls, we
# rebuild the same split on an array of "defect type" strings to
# recover this information without re-loading any images.
# ============================================================

all_defect_types = []
for d in OK_DIRS:
    for f in sorted(d.glob("*.png")):
        all_defect_types.append("good")

for d in NOT_OK_DIRS:
    defect_name = d.name
    for f in sorted(d.glob("*.png")):
        all_defect_types.append(defect_name)

all_defect_types = np.array(all_defect_types)

indices = np.arange(len(all_paths))

idx_trainval, idx_test = train_test_split(
    indices, test_size=0.15, stratify=all_labels, random_state=SEED
)
idx_train, idx_val = train_test_split(
    idx_trainval, test_size=0.18, stratify=all_labels[idx_trainval], random_state=SEED
)

assert len(idx_test) == len(X_test), "Index split doesn't match X_test size — check split parameters match Cell 5 exactly"

test_defect_types = all_defect_types[idx_test]

print(f"Test set defect-type breakdown: {pd.Series(test_defect_types).value_counts().to_dict()}")

results_df = pd.DataFrame({
    "true_label": y_test,
    "defect_type": test_defect_types,
    "pred_default": y_test_pred_default,
    "pred_tuned": y_test_pred_tuned,
    "prob": y_test_prob,
})

defect_only = results_df[results_df["defect_type"] != "good"]

breakdown = defect_only.groupby("defect_type").apply(
    lambda g: pd.Series({
        "n_samples": len(g),
        "caught_default": (g["pred_default"] == 1).sum(),
        "missed_default": (g["pred_default"] == 0).sum(),
        "caught_tuned": (g["pred_tuned"] == 1).sum(),
        "missed_tuned": (g["pred_tuned"] == 0).sum(),
        "mean_prob": g["prob"].mean(),
    })
).round(3)

print("\nPer-defect-type breakdown (test set):")
print(breakdown.to_string())

fig, ax = plt.subplots(figsize=(8, 5))
x = np.arange(len(breakdown))
width = 0.35
ax.bar(x - width/2, breakdown["caught_default"] / breakdown["n_samples"], width,
       label="Caught (default thr. 0.5)", color="lightcoral")
ax.bar(x + width/2, breakdown["caught_tuned"] / breakdown["n_samples"], width,
       label=f"Caught (tuned thr. {CHOSEN_THRESHOLD})", color="seagreen")
ax.set_xticks(x)
ax.set_xticklabels(breakdown.index)
ax.set_ylabel("Fraction of defects caught (recall)")
ax.set_title("Recall by Defect Type — Default vs Tuned Threshold")
ax.legend()
ax.set_ylim(0, 1.05)
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
